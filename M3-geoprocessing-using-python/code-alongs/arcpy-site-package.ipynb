### 1. Importing ArcPy

So far you have used functions that come with Python, such as `print()` and `len()`, and functions you wrote yourself. Python can do much more once you bring in code that someone else has written.

* A **module** is a Python file that contains functions, classes, and variables you can reuse.
* A **package** is a collection of related modules organized together.
* A **site package** is a package installed into your Python environment that adds new functionality to Python. ArcPy is a site package that is installed with ArcGIS Pro.

To use a package, you **import** it with the `import` keyword. Once it is imported, you access anything inside it with **dot notation**: the package name, a dot `.`, then the name of the item, such as `arcpy.GetInstallInfo`.

ArcPy is only available in the ArcGIS Pro Python environment. If this notebook's kernel is not set to that environment, `import arcpy` raises a `ModuleNotFoundError`. The first import can take several seconds.

```python
# Import the ArcPy site package
import arcpy
```

`arcpy.GetInstallInfo()` is an ArcPy function that returns a dictionary of details about the ArcGIS Pro installation on this computer. Because the return value is a dictionary, you can loop over it with `.items()`.

```python
# Store the dictionary returned by GetInstallInfo()
install_info = arcpy.GetInstallInfo()
print(f"type(install_info): {type(install_info)}")

# Loop over the dictionary to print each key and its value
print("Contents of Install Info dictionary:")
for key, value in install_info.items():
    print(f"\t{key}: {value}")
```

### 4. Classes, properties, and methods

A class is a blueprint for creating objects. A **class** defines what data its objects store and what actions they can perform. An **object** is one specific value created from a class, and each object stores its own data. Creating an object looks like calling a function: the class name, then parentheses with arguments inside.

* A **property** is a piece of data stored in an object. You access it with dot notation and no parentheses, such as `utm_sr.name`.
* A **method** is a function that belongs to an object. You call it with dot notation and parentheses, such as `utm_sr.exportToString()`. You have already used methods: `.upper()` is a string method and `.append()` is a list method.

The ArcGIS Pro Python reference provides a list of [ArcPy Classes](https://doc.esri.com/en/arcgis-pro/latest/arcpy/classes/alphabetical-list-of-arcpy-classes.html) and includes the description, syntax and code samples for each class.

### 5. SpatialReference class
The `SpatialReference` class describes a coordinate system. You can create a `SpatialReference` object from a coordinate system's **factory code**, also called its **WKID** (well-known ID), a number that uniquely identifies a coordinate system. You can also create one from the coordinate system's name.

The [SpatialReference class documentation](https://doc.esri.com/en/arcgis-pro/latest/arcpy/classes/spatialreference.html) lists all of its properties and methods. It also includes information and factory codes for all geographic coordinate systems as well as all projected coordinate systems.

```python
# Example 1 - Create a SpatialReference object from a factory code

# WKID 26913 = NAD 1983 UTM Zone 13N
utm_sr = arcpy.SpatialReference(26913)
# utm_sr stores a SpatialReference object
print(f"type(utm_sr): {type(utm_sr)}")

# Properties use dot notation with no parentheses
print(f"Name: {utm_sr.name}")
print(f"Type: {utm_sr.type}")
print(f"Factory code: {utm_sr.factoryCode}")
print(f"Linear unit: {utm_sr.linearUnitName}")
```

```python
# Example 2 - Create a SpatialReference object from a coordinate system name

# WGS 1984
wgs_sr = arcpy.SpatialReference("WGS 1984")
# wgs_sr stores a SpatialReference object

# Printing out properties
print(f"Name: {wgs_sr.name}")
print(f"Type: {wgs_sr.type}")
print(f"Factory code: {wgs_sr.factoryCode}")

# A geographic coordinate system uses angular units, not linear units
print(f"Angular unit: {wgs_sr.angularUnitName}")
```

```python
# Methods use dot notation with parentheses. exportToString() returns the
# full definition of the coordinate system as a string
# We will use variable utm_sr (SpatialReference object) from Example 1
utm_string = utm_sr.exportToString()
print(f"utm_string: {utm_string}")
```

### 6. env class
In ArcGIS Pro, **environment settings** are settings that apply to every tool you run, such as where data is stored and whether outputs can be replaced. ArcPy stores these settings as properties of `arcpy.env`, an object created from the `env` class. Setting a property on `arcpy.env` works like assigning a value to a variable, and every tool and function that runs afterward uses the new value.

Two settings appear in almost every ArcPy script:

* `workspace`: the default location of input and output datasets. Once the workspace is set, you can refer to a dataset in it by its name instead of its full path.
* `overwriteOutput`: when `True`, a tool replaces an output dataset that already exists. When `False`, running a tool again with the same output name raises an error.

The [env class documentation](https://doc.esri.com/en/arcgis-pro/latest/arcpy/classes/env.html) lists all of its properties.

```python
# Set the workspace to the data folder. Update this path to match the
# location of the data folder on your computer. Windows paths use a raw
# string
arcpy.env.workspace = r"c:\gist\data"
print(f"arcpy.env.workspace: {arcpy.env.workspace}")

# Allow tools to replace outputs that already exist
arcpy.env.overwriteOutput = True
print(f"arcpy.env.overwriteOutput: {arcpy.env.overwriteOutput}")
```

⚠️ **Important**: Anything in ArcPy that takes a dataset, whether a function, a tool, or a class, accepts it in two ways:

* A **full path** includes every folder from the drive down to the dataset, such as `r"C:\gist\data\pourpoints.shp"`. ArcPy uses the full path exactly as written and ignores the workspace.
* A **dataset name** without a path, such as `"pourpoints.shp"`, is looked up in the workspace. ArcPy joins `arcpy.env.workspace` and the name to build the full path for you. When a tool's output is given as a name, the output is also written to the workspace.

Setting the workspace once lets you refer to every dataset in that folder by its name alone. This keeps your code shorter and easier to update: if the data moves, you only change the workspace.
