### 1. Creating and printing strings

Strings generally use single quotes or double quotes. Either works, remember to be consistent. Use double quotes when your text itself contains an apostrophe. Triple quotes let a string span multiple lines. 

```python
# Strings can use single or double quotes
station_name = "Bear Butte 1"
checker = 'A Thomas'

print(station_name)
print(checker)

# Use double quotes when your string contains an apostrophe
note = "Bear Butte's checkpoint needs a re-check next cycle"
print(note)

# Triple quotes let a string span multiple lines
field_notes = """Station: Bear Butte 1
Quality: Fair
Checked by: B Stewart"""
print(field_notes)
```

Escape sequences insert characters that are hard to type directly. Unicode escapes (`\uXXXX`) insert a specific character by its code point, useful for accented letters or other special characters.  
Refer to [Unicode 18.0 Character Code Charts](https://www.unicode.org/Public/18.0.0/charts/) for a list of Unicode codes.

```python
# Escape sequences
print('Checker said, \'looks good\'')      # escaped single quote
print("Checker said, \"looks good\"")      # escaped double quote
print("Line 1\nLine 2")                    # \n starts a new line
print("C:\\Workspace\\GIST\\")             # \\ inserts a single backslash

# Unicode escapes: \u takes 4 hex digits (code points up to U+FFFF)  
print("Ca\u00f1on del Oro Wash")           # \u00f1 -> n with a tilde

# Unicode escapes: \U takes 8 hex digits, like most emoji
print("Field check complete \U0001F44D")   # \U0001F44D -> thumbs up emoji 
```

### 2. Indexing and slicing

Every character in a string has an **index** — its position, starting at 0. You access a single character with its index (`s[0]`), or a range of characters with a **slice** (`s[start:end]`, where `end` is not included). Negative indices count from the end (`-1` is the last character).

```python
check_date = "2021-12-21"

print(check_date[0])       # index 0 -> first character
print(check_date[-1])      # index -1 -> last character
print(check_date[0:4])     # indices 0-3 -> first 4 characters (the year)
print(check_date[5:7])     # indices 5-6 -> the month
print(check_date[-2:])     # indices -2 to end -> last 2 characters (the day)
```

### 3. Building strings with f-strings

f-strings are the modern, recommended way to build a string out of variables: put an `f` before the quotes and drop variables directly inside `{}`. You may also come across two older styles in existing code or AI-generated code: **concatenation** (joining strings together with `+`) and `.format()` method. You don't need to use them yourself, but you should be able to recognize what they're doing when you see them.

```python
# We will use variable station_name from cell 1: station_name = "Bear Butte 1"
# We will use variable checker from cell 1: checker = 'A Thomas'
# New variable that describes a quality
qual = "Excellent"

# f-strings demo
summary = f"{station_name} - quality: {qual} (checked by {checker})"
print(summary)

# Older styles that do the same thing. Recognize these, no need to write them
# concatenation using +
print(station_name + " - quality: " + qual + " (checked by " + checker + ")")
# .format() method
print("{} - quality: {} (checked by {})".format(station_name, qual, checker))

# Two variables that store floats
x_coord = 610124.0815
y_coord = 4905101.303

# A colon after the variable inside {} formats the value
# .2f rounds x_coord/y_coord to 2 decimal places
coord_summary = f"Station at ({x_coord:.2f}, {y_coord:.2f})"
print(coord_summary)
```

### 4. Writing file paths

File paths are stored as strings in Python. Windows file paths use backslashes (`\`) to separate folders. However, in Python, a backslash indicates an escape sequence like `\n` or `\t`. If a folder or file name starts with an n or t, it can accidentally form a real escape sequence and the path will be corrupted. There are three common ways to write a windows-based path correctly, and a fourth once f-strings get involved.

```python
# Windows path to a made-up shapefile. Note the \n and \t 
print("C:\workspace\gist\new_data\trees.shp")

# Three ways to write the same Windows file path correctly

# forward slashes -- Python (and Windows) both accept these
print("C:/workspace/gist/new_data/trees.shp")

# escaped backslashes -- each \\ becomes one \
print("C:\\workspace\\gist\\new_data\\trees.shp")

# raw string -- the r prefix turns off escape processing
print(r"C:\workspace\gist\new_data\trees.shp")

# f-strings are not raw by default. 
# Combine both prefixes (rf"..." or fr"...") to get raw text

# raw string, so backslashes are literal characters
folder = r"C:\workspace\gist\new_data"

# plain string -- no backslashes, no special handling needed
trees_shp = "trees.shp"

# rf-string: {} still interpolates, and the \ between them stays literal
full_path = rf"{folder}\{trees_shp}"
print(full_path)  # C:\workspace\gist\new_data\trees.shp
```

### 5. Cleaning and checking text with string methods

Real-world GIS data is messy. Strings may include inconsistent capitalization, stray whitespace, the odd typo in a name field. String methods such as `.strip()`, `.upper()`/`.lower()`/`.title()`, `.replace()`, `.startswith()`/`.endswith()` can help with data cleanup. The `in` operator can be used for substring checks.

```python
# Messy field values are extremely common in real GIS data
raw_qual = "  exceLLent  "

print(raw_qual.strip())                 # remove leading/trailing whitespace
print(raw_qual.strip().upper())         # then convert to uppercase
print(raw_qual.strip().title())         # or title case

# Output of methods can also be assigned to variables
lower_qual = raw_qual.strip().lower()   # converting to lower case
print(lower_qual)

print(lower_qual.startswith("e"))       # True
print(lower_qual.startswith("E"))       # False -- Python is case-sensitive

print(lower_qual.endswith("ent"))       # True
print(lower_qual.endswith("ENT"))       # False -- Python is case-sensitive

print("ex" in lower_qual)               # True -- substring check
print("Ex" in lower_qual)               # False -- Python is case-sensitive

# Fixing a typo in station name
typo_name = "Bear Buttes 1"
print(typo_name.replace("Buttes", "Butte"))  # Replaces the first occurrence
```
