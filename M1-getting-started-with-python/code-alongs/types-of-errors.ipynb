### 1. Syntax errors

Python reads your code before it runs any of it, checking that it follows the rules of the language. These rules are called **syntax**, and this reading step is called **parsing**. Just as a sentence with a missing word can be hard to understand, code that breaks a syntax rule can't be understood by Python at all.

A **syntax error** happens when code breaks one of these rules. Because Python parses the whole cell before running it, a syntax error stops *every* line in the cell from running.

A **literal** is a value typed directly into code, like `42` or `"Elk 1"`. When you create a variable, the name goes on the left of `=` and the value on the right. Swapping the two sides is a syntax error, because Python can't store a value inside a literal.

```python
# The value is on the left and the variable name on the right: backwards
42 = the_answer
```

How to read a syntax error message:

* The line number shows which line Python was reading when it got stuck. Line numbers count every line in the cell, including comments.
* The `^` marker points to where on that line the problem is
* The last line starts with `SyntaxError:` followed by a short description

Python sometimes suggests a fix at the end of the message. Suggestions are guesses and can be wrong: here, the real fix is to swap the two sides of `=`, as in `the_answer = 42`.

Another common syntax error is forgetting to close a string's quotes.

```python
# The closing quote is missing
station_name = "Elk 1
```

Forgetting to close a parenthesis `()` is also a syntax error. In the cell below, lines 2 and 4 are correct, but neither one runs: Python found the syntax error on line 3 while parsing, so it never started running the cell. Notice that nothing is printed above the error message.

```python
# Lines 2 and 4 are correct but never run: line 3 is missing a )
print("Checking station Elk 1")
print("Station quality: Fair"
print("Check complete")
```

If the line missing its closing parenthesis is the *last* line in a cell, the message is less helpful: `SyntaxError: incomplete input`. It means the same thing. Python reached the end of the cell while still waiting for something to be closed.

### 2. Exceptions

When code follows all the syntax rules, Python starts running it line by line. An **exception** is an error that happens *while* the code is running: Python reaches a line it can't carry out, stops, and reports what went wrong. Unlike a syntax error, every line *before* the problem line has already run.

The report Python shows is called a **traceback**. A traceback can look long, but read it from the bottom up:

* The last line gives the exception's name, called its **type** (for example, `NameError`), followed by a message describing the problem
* Above it, an arrow `---->` points to the line of code that caused the exception

Each type of exception describes a different kind of problem. The three below are among the most common for beginners.

**NameError**: Python doesn't recognize a name. Usually the variable name is misspelled, uses different capitalization (names are case-sensitive), or the cell that creates it hasn't been run yet.

```python
station_name = "Elk 1"
print(f"station_name: {station_name}")
```

**TypeError**: an operation was used on a data type that doesn't support it. A common case: a number read from a file arrives as a string, and you try to do math with it. When `+` has a string on its left, Python tries to **concatenate** (join end to end) two strings, and a number on the right isn't a string.

```python
elevation_text = "5320"  # read from a file, so it's a string
raised_ft = elevation_text + 100
```

**ValueError**: a function received the right data type, but a value it can't use. `float()` accepts strings, but only strings that look like a number. Missing values in data files are often written as text like `"N/A"`, which `float()` can't convert.

```python
y_text = "N/A"  # a missing coordinate in a CSV file
y_value = float(y_text)
```

Python has many other types of exceptions, and you'll meet some of them later. Every traceback is read the same way: start with the last line.

### 3. Logical errors

The hardest errors to find are the ones Python doesn't report. A **logical error** is a mistake in the code's thinking rather than its syntax: the code follows every rule and runs to the end, but the result is wrong. There's no error message and no traceback, so the only way to catch a logical error is to check whether the output makes sense.

```python
buffer_miles = 10
buffer_feet = 52800  # 10 miles in feet (1 mile = 5280 feet)

# Runs without an error, but prints the wrong variable
print(f"Using {buffer_feet}-mile buffer")
```

Python has no way of knowing that a 52800-mile buffer is wrong. The code follows every rule; it just uses `buffer_feet` where `buffer_miles` was meant.

Order of operations is another common source of logical errors. `*` and `/` are calculated before `+` and `-`, so an average written without parentheses gives a wrong answer without any error.

```python
elk_1_elev_ft = 5320
elk_2_elev_ft = 5480

# Runs without an error, but / is calculated before +
avg_elev_ft = elk_1_elev_ft + elk_2_elev_ft / 2
print(f"avg_elev_ft: {avg_elev_ft}")
```

A quick sanity check catches this one: an average can't be larger than both of the values being averaged. The fix is `(elk_1_elev_ft + elk_2_elev_ft) / 2`.
