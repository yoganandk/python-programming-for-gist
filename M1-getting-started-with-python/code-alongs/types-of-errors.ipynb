# Types of Errors

Every programmer, from beginner to expert, sees error messages every day. An error message is not a sign that you have failed. It's Python telling you what went wrong and, often, where. In this notebook, you will learn to recognize the three types of errors you will run into when writing Python, so that when your code stops working, or quietly gives you a wrong answer, you know what kind of problem you are dealing with.

* Recognize an exception and read its traceback to find the exception type and the line that caused it
* Recognize a logical error, which gives no error message at all

### 1. Syntax errors

Python reads your code before it runs any of it, checking that it follows the rules of the language. These rules are called **syntax**, and this reading step is called **parsing**. Just as a sentence with a missing word can be hard to understand, code that breaks a syntax rule cannot be understood by Python at all.

A **syntax error** happens when code breaks one of these rules. Because Python parses the whole cell before running it, a syntax error stops *every* line in the cell from running.

How to read a syntax error message:
* The line number shows which line Python was reading when it got stuck. Line numbers count every line in the cell, including comments. In VS Code, you can click the line number to jump to that line.
* The `^` marker points to where on that line the problem is.
* The last line starts with `SyntaxError:` followed by a short description.

Run each of the three examples below, one at a time, and read its error message.

Syntax Errors - Example 1

```python
# Example 1
# The value is on the left and the variable name on the right: backwards
42 = the_answer
print(the_answer)
```

In Example 1, the description says: `cannot assign to literal here. Maybe you meant '==' instead of '='?`

A **literal** is a value typed directly into code, like `42` or `"Elk 1"`. When you create a variable, the name goes on the left of `=` and the value on the right. Swapping the two sides asks Python to store a value inside a literal, which is not possible.

Python sometimes suggests a fix at the end of the message. Suggestions are guesses and can be wrong: here, the real fix is to swap the two sides of `=`, as in `the_answer = 42`.

Notice that `print(the_answer)` on line 4 did not run either, even though it is correct.

```python
# Example 1 - Corrected Code
the_answer = 42
print(the_answer)
```

Syntax Errors - Example 2

```python
# Example 2
# The closing quote is missing
print("My name is ...)
print("John Doe")
```

In Example 2, the description says: `unterminated string literal (detected at line 3)`

`unterminated` means "not closed." A string typed directly into code is a string literal, so `unterminated string literal` means a string is missing its closing quote.

`(detected at line 3)` tells you which line the problem is on, and the `^` points to where the string starts, which helps you find which quote needs a partner.

Notice that `print("John Doe")` on line 4 did not run either.

```python
# Example 2 - Corrected Code
print("My name is ...")
print("John Doe")
```

Syntax Errors - Example 3

```python
# Example 3
# The closing parenthesis is missing
print("My name is ..."
print("John Doe")
```

In Example 3, the description says: `'(' was never closed`

The `(` after `print` opens the function call, and every opening parenthesis needs a closing one `)`. The `^` points to the `(` that is missing its partner.

Notice that `print("John Doe")` on line 4 did not run either.

If the line missing its closing parenthesis is the last line in a cell, the message is less helpful: `incomplete input`. It means the same thing. Python reached the end of the cell while still waiting for something to be closed.

```python
# Example 3 - Corrected Code
print("My name is ...")
print("John Doe")
```

### 2. Exceptions

When code follows all the syntax rules, Python starts running it line by line. An **exception** is an error that happens *while* the code is running: Python reaches a line it cannot carry out, stops, and reports what went wrong. This is called **raising** an exception. Unlike a syntax error, every line *before* the problem line has already run.

The report Python shows is called a **traceback**. A traceback can look long, but read it from the bottom up.

How to read a traceback:
* The last line gives the name of the exception, called its **exception type** (for example, `NameError`), followed by a message describing the problem.
* Above it, an arrow `---->` points to the line of code that caused the exception. In VS Code, you can click the line number to jump to that line.

Each exception type describes a different kind of problem. The three examples below show some of the most common exception types for beginners. Run each one, one at a time, and read its traceback.

Exceptions - Example 1 - NameError

```python
# Example 1 - NameError
station_name = "Elk 1"  # note that variable is all lowercase
print("Looking up station...")
print("station_name:", Station_Name)  # variable has capital S and N
```

In Example 1, the last line of the traceback says: `NameError: name 'Station_Name' is not defined`

A **NameError** means Python does not recognize a name. Usually the variable name is misspelled, uses different capitalization (names are case-sensitive), or the cell that creates it has not been run yet. Here, the variable was created as `station_name`, but line 4 uses `Station_Name`.

The arrow `---->` points to line 4, the line that caused the exception.

Notice that `Looking up station...` printed above the traceback. Lines 2 and 3 ran, and the exception stopped the cell at line 4.

```python
# Example 1 - Corrected Code
station_name = "Elk 1"  # note that variable is all lowercase
print("Looking up station...")
print("station_name:", station_name)  # variable is all lowercase
```

Exceptions - Example 2 - TypeError

```python
# Example 2 - TypeError
elevation_text = "5320"  # elevation is stored as a string
raised_ft = elevation_text + 100
print("raised_ft:", raised_ft)
```

In Example 2, the last line of the traceback says: `TypeError: can only concatenate str (not "int") to str`

A **TypeError** means an operation was used on a data type that does not support it. `elevation_text` holds `"5320"` in quotes, so it is a string, not a number. When `+` has a string on its left, Python tries to **concatenate** (join end to end) two strings, but `100` is an `int`.

The arrow `---->` points to line 3, the line that caused the exception. Line 4 never ran.

The fix is to convert the string to a number with `int()` before doing math.

```python
# Example 2 - Corrected Code
elevation_text = "5320"  # elevation is stored as a string
raised_ft = int(elevation_text) + 100  # string is converted to an integer
print("raised_ft:", raised_ft)
```

Exceptions - Example 3 - ValueError

```python
# Example 3 - ValueError
y_text = "N/A"  # a missing value, stored as text
y_value = float(y_text)
print("y_value:", y_value)
```

In Example 3, the last line of the traceback says: `ValueError: could not convert string to float: 'N/A'`

A **ValueError** means a function received the right data type but a value it cannot use. `float()` accepts strings, but only strings that look like a number. `"N/A"` is a string, but it does not look like a number, so `float()` cannot convert it.

The arrow `---->` points to line 3, the line that caused the exception. Line 4 never ran.

The fix is to give `float()` a string that looks like a number.

```python
# Example 3 - Corrected Code
y_text = "4905101.303"  # a y coordinate, stored as a string
y_value = float(y_text)  # string is converted to a float
print("y_value:", y_value)
```

### 3. Logical errors

The hardest errors to find are the ones Python does not report. A **logical error** is a mistake in the logic of the code: the code follows every syntax rule and runs to the end without raising an exception, but the result is wrong.

Unlike syntax errors and exceptions, a logical error produces no error message and no traceback. The only way to catch one is to check whether the output makes sense.

Run each example below, one at a time, and check whether its output makes sense.

Logical Errors - Example 1

```python
# Example 1
buffer_miles = 10
buffer_feet = 52800  # 10 miles in feet (1 mile = 5280 feet)
print("Using", buffer_feet, "mile buffer")
```

In Example 1, the cell runs without an error and prints: `Using 52800 mile buffer`

A 52800 mile buffer is more than twice the distance around the Earth. Python has no way of knowing that is wrong. The code follows every rule; it just uses `buffer_feet` where `buffer_miles` was meant.

```python
# Example 1 - Corrected Code
buffer_miles = 10
buffer_feet = 52800  # 10 miles in feet (1 mile = 5280 feet)
print("Using", buffer_miles, "mile buffer")
```

Logical Errors - Example 2

```python
# Example 2
elk_1_elev_ft = 5320
elk_2_elev_ft = 5480
avg_elev_ft = elk_1_elev_ft + elk_2_elev_ft / 2
print("avg_elev_ft:", avg_elev_ft)
```

In Example 2, the cell runs without an error and prints: `avg_elev_ft: 8060.0`

An average cannot be larger than both of the values being averaged, so this result is wrong. Python follows the order of operations: `/` is calculated before `+`, so the code calculates `5320 + (5480 / 2)` instead of `(5320 + 5480) / 2`. The fix is to add parentheses so the addition happens first.

```python
# Example 2 - Corrected Code
elk_1_elev_ft = 5320
elk_2_elev_ft = 5480
avg_elev_ft = (elk_1_elev_ft + elk_2_elev_ft) / 2
print("avg_elev_ft:", avg_elev_ft)
```
