### 1. Data types in Python

Every value in Python has a **data type**, which determines what kind of value it is and what you can do with it. For example, you can do math with a number, but not with text.

Python has four basic data types:

* `int` (integer): a whole number, positive or negative, with no decimal point
* `float` (floating-point number): a number with a decimal point
* `str` (string): text, written inside single or double quotes
* `bool` (boolean): either `True` or `False`, always capitalized and never in quotes

```python
5320            # int
-42             # int
610124.0815     # float
0.3048          # float
"Elk 1"         # str
'Bear Butte 1'  # str
True            # bool
False           # bool
```

💬 **Comments**: Text that follows a `#` is a comment, a note for people reading the code. Python ignores comments. Above, each comment labels the data type of the value on its line.

⚠️ **Quotes decide the data type**: A number or word written inside quotes is a string, even if it looks like a number or a boolean.

```python
5320     # int
"5320"   # str
True     # bool
"True"   # str
```

Python also has other data types: lists, tuples, ranges, and dictionaries that store collections of values, and `None` represents "no value." These are covered in later notebooks.

### 2. Variables

A **variable** is a name that stores a value so you can use it later in your code. You create a variable with `=`, the **assignment operator**. In Python, `=` does not mean "equals" the way it does in math. It means "store the value on the right in the name on the left."

Assigning a new value to an existing variable replaces the old value; this is called **reassignment**.

Here are some examples. Running the following cell stores the values, but displays nothing. Displaying values in the output is covered in Section 3.

````python
# Store the string "Elk 1" in a variable named station_name
station_name = "Elk 1"

# Store the integer 5320 in a variable named elevation_ft
elevation_ft = 5320

# Store the float 610124.0815 in a variable named x_coord
x_coord = 610124.0815

# Store the boolean False in a variable named is_checked
is_checked = False

# Reassignment: is_checked now stores True
is_checked = True
````

⚠️ **Rules for naming variables**:

* Names can contain letters, numbers, and underscores
* Names cannot start with a number
* Names cannot contain spaces or other special characters
* Names cannot be words Python reserves for itself, such as `True`, `False`, `if`, or `for`
* Names are **case-sensitive**: `station_name` and `Station_Name` are two different variables

**Convention**: [PEP 8](https://peps.python.org/pep-0008/), Python's official style guide, recommends writing variable names as lowercase words joined by underscores, a style known as **snake_case**.

````python
# Valid variable names

# snake_case: lowercase words joined by underscores (the convention)
checker_name = "B Stewart"
station_2_name = "Bear Butte 1"   # numbers are fine after the first letter

# Single letters: valid, but they do not say what they store
a = "B Stewart"   # is a a checker? a station? a date?
b = 5320          # 5320 what? feet? meters? an ID?
i = 1             # i is a common exception, often used as a counter

# Other styles: valid, but not the convention for variable names
stationName = "Elk 1"       # camelCase: used in other languages
StationName = "Elk 1"       # PascalCase: used for classes in Python
FEET_TO_METERS = 0.3048     # UPPER_CASE: used for values that never change
_station_name = "Elk 1"     # leading underscore: marks as "internal use only"

# Case-sensitive: these are three different variables
station_name = "Elk 1"
Station_Name = "Elk 2"
STATION_NAME = "Elk 3"
````

### 3. Built-in functions

A **function** is a named, ready-made block of code that performs a task. Python comes with many **[built-in functions](https://docs.python.org/3/builtins/functions.html)** that are always available. You **call** a function by writing its name followed by parentheses. The value inside the parentheses is the function's **argument**. Many functions hand back a result, called the function's **return value**.

`print()` displays its argument below the cell. The argument can be any value or a variable.

Quotes change what `print()` displays: without quotes, Python treats a word as a variable name and prints the value stored in it; with quotes, Python treats it as a string and prints the text exactly as written.

````python
# print() Example 1

# print() displays the value it receives
print("Elk 1")  # Value is a string "Elk 1"
````

````python
# print() Example 2

# print() displays the value it receives
print(5320)  # Value is an integer 5320
````

````python
# print() Example 3

# Store the string "Elk 1" in a variable named station_name
station_name = "Elk 1"

# No quotes: Python prints the value stored in the variable
print(station_name)

# Quotes: Python prints the text exactly as written
print("station_name")
````

`print()` can display several values at once when they are separated by commas. It prints them on one line with a space between each. This is useful for printing a sentence built from values, or for putting a string label before a variable to show which value is which.

````python
# Store the string "Elk 1" in a variable named station_name
station_name = "Elk 1"

# Store the integer 5320 in a variable named elevation_ft
elevation_ft = 5320

# A string label, a comma, then the variable
# print() adds a space between the label and the value
print("station_name:", station_name)

# Several values: print() adds a space between each one
print(station_name, "is at", elevation_ft, "feet")
````

`type()` returns the data type of its argument. Its return value can be stored in a variable, like any other value.

````python
# type() Example 1 - string data type

# Store the string "Elk 1" in a variable named station_name
station_name = "Elk 1"

# Call type() with station_name as its argument, and store the return value
station_name_type = type(station_name)

# Print the data type
print("station_name_type:", station_name_type)

# Alternatively, call type() directly inside print(): type() runs first,
# then print() displays its return value
print("type(station_name):", type(station_name))
````

📝 **Note**: `type()` reports a data type in the form `<class 'str'>`. In Python, every data type is a **class**: a blueprint that defines what a kind of value is and what it can do. For now, read that simply as `str`.

````python
# type() Example 2 - integer data type

# Store the integer 5320 in a variable named elevation_ft
elevation_ft = 5320

# Call type() with elevation_ft as its argument, and store the return value
elevation_ft_type = type(elevation_ft)

# Print the stored data type
print("elevation_ft_type:", elevation_ft_type)

# Alternatively, call type() directly inside print(): type() runs first,
# then print() displays its return value
print("type(elevation_ft):", type(elevation_ft))
````

````python
# type() Example 3 - float data type

# Store the float 610124.0815 in a variable named x_coord
x_coord = 610124.0815

# Call type() with x_coord as its argument, and store the return value
x_coord_type = type(x_coord)

# Print the stored data type
print("x_coord_type:", x_coord_type)

# Alternatively, call type() directly inside print(): type() runs first,
# then print() displays its return value
print("type(x_coord):", type(x_coord))
````

````python
# type() Example 4 - boolean data type

# Store the boolean False in a variable named is_checked
is_checked = False

# Call type() with is_checked as its argument, and store the return value
is_checked_type = type(is_checked)

# Print the stored data type
print("is_checked_type:", is_checked_type)

# Alternatively, call type() directly inside print(): type() runs first,
# then print() displays its return value
print("type(is_checked):", type(is_checked))
````

### 4. Converting between data types

The data type of a value controls how it behaves. A number stored as a string looks the same as the number when printed, but Python treats it as text.

````python
# Store the integer 5320
elevation_ft = 5320      # This is an integer

# Store the string "5320"
elevation_text = "5320"  # This is a string

# Both print the same...
print("elevation_ft:", elevation_ft)
print("elevation_text:", elevation_text)

# ...but they are different data types...
print("type(elevation_ft):", type(elevation_ft))
print("type(elevation_text):", type(elevation_text))

# ...so they behave differently
print("elevation_ft * 2:", elevation_ft * 2)      # performs multiplication
print("elevation_text * 2:", elevation_text * 2)  # repeats the string
````

Values do not always arrive as the data type you need. For example, every value read from a CSV or text file arrives as a string, even a coordinate or an elevation. **Converting** a value turns it into a different data type.

The built-in functions `int()`, `float()`, and `str()` each take one argument and return it converted to that data type.

````python
# Example 1 - float()

# A coordinate read from a CSV file arrives as a string
x_text = "610124.0815"
print("x_text:", x_text)
print("type(x_text):", type(x_text))

# float() converts a string that looks like a decimal number into a float
x_coord = float(x_text)
print("x_coord:", x_coord)
print("type(x_coord):", type(x_coord))
````

````python
# Example 2 - int()

# An elevation read from a text file arrives as a string
elevation_text = "5320"
print("elevation_text:", elevation_text)
print("type(elevation_text):", type(elevation_text))

# int() converts a string of digits into an int
elevation_ft = int(elevation_text)
print("elevation_ft:", elevation_ft)
print("type(elevation_ft):", type(elevation_ft))
````

````python
# Example 3 - str()

# Store the integer 5320 in a variable named elevation_ft
elevation_ft = 5320
print("elevation_ft:", elevation_ft)
print("type(elevation_ft):", type(elevation_ft))

# str() converts any value into a string
elevation_str = str(elevation_ft)
print("elevation_str:", elevation_str)
print("type(elevation_str):", type(elevation_str))
````

### 5. Math with numbers

Once values are stored as numbers, not strings, you can calculate with them. Python does math with `int` and `float` values using **arithmetic operators**: `+` (add), `-` (subtract), `*` (multiply), and `/` (divide).

The data type of the result follows two rules:

* If any value in the calculation is a `float`, the result is a `float`
* Division with `/` always returns a `float`, even when the result is a whole number

````python
# Variable stores an integer
elevation_ft = 5320

# int + int returns an int
raised_ft = elevation_ft + 100
print("raised_ft:", raised_ft)
print("type(raised_ft):", type(raised_ft))

# int * float returns a float, because one of the values is a float
elevation_m = elevation_ft * 0.3048  # 1 foot = 0.3048 meters
print("elevation_m:", elevation_m)
print("type(elevation_m):", type(elevation_m))

# / always returns a float, even when the result is a whole number
half_ft = elevation_ft / 2
print("half_ft:", half_ft)
print("type(half_ft):", type(half_ft))
````

Python follows the usual order of operations: `*` and `/` are calculated before `+` and `-`. Parentheses `()` control which part is calculated first.

````python
# Variables store integers
elk_1_elev_ft = 5320
elk_2_elev_ft = 5480

# Without parentheses, / runs first: 5320 + (5480 / 2)
avg_wrong_ft = elk_1_elev_ft + elk_2_elev_ft / 2
print("Without parentheses:", avg_wrong_ft)

# With parentheses, + runs first: (5320 + 5480) / 2
avg_elev_ft = (elk_1_elev_ft + elk_2_elev_ft) / 2
print("With parentheses:", avg_elev_ft)
````

Math with floats often produces results with many decimal places. `int()` converts a `float` to an `int` by removing everything after the decimal point. It does **not** round.

To round to the nearest whole number, use the built-in `round()` function instead. `round()` also accepts a second argument, separated by a comma, for the number of decimal places to keep.

````python
# Variable stores a float
elevation_m = 1621.536

# int() removes the decimal part
elevation_m_int = int(elevation_m)
print("elevation_m_int:", elevation_m_int)
print("type(elevation_m_int):", type(elevation_m_int))

# round() rounds to the nearest whole number
elevation_m_round = round(elevation_m)
print("elevation_m_round:", elevation_m_round)
print("type(elevation_m_round):", type(elevation_m_round))

# A second argument sets the number of decimal places to keep
elevation_m_2dp = round(elevation_m, 2)
print("elevation_m_2dp:", elevation_m_2dp)
print("type(elevation_m_2dp):", type(elevation_m_2dp))
````
