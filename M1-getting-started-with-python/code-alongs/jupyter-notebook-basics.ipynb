After a code cell runs, a number in square brackets, such as `[1]`, appears to the left of the cell, beside its last line. This is the **execution count**: it records the order in which cells were run. The first cell you run gets `[1]`, the next gets `[2]`, and so on. A cell that has not been run yet shows no number.

The execution count follows the order cells were *run*, not the order they appear on the page. Running a cell again gives it a new, higher number. Section 3 explains why that order matters.

### 3. The kernel

When you run a code cell, the notebook does not run the Python code itself. It sends the code to the **kernel**: a Python program running in the background that carries out the code and sends the output back to the notebook. The kernel this notebook uses is shown in the top-right corner of the notebook and should read `arcgispro-py3-env`, the copy of Python that is installed with ArcGIS Pro. This is the kernel you selected at the start of the notebook.

The kernel also *remembers* what earlier cells did. Anything a cell creates stays in the kernel's memory, and cells run later can use it. This is why **run order matters**: a cell that relies on something another cell created only works if that other cell has already been run. It works like a geoprocessing workflow: you can't clip roads to a study area until the step that creates the study area boundary has run. You will see this with real code once you start storing values in Python.

**Restarting** the kernel stops it and starts a fresh one with an empty memory: everything earlier cells created is forgotten. Restarting does *not* clear the screen. Old outputs and execution counts stay visible, even though the new kernel knows nothing about them, and the next cell you run starts again at `[1]`.

Three buttons in the toolbar along the top of the notebook control this:

* **Restart**: restarts the kernel
* **Run All**: runs every cell from top to bottom
* **Clear All Outputs**: removes the output from every cell

**Run All does not restart the kernel.** If the kernel still remembers something from an earlier run, a cell can appear to work even though it would fail from a fresh start. Leftover outputs can hide the same problem. To check that a notebook really works from top to bottom, do a **clean rerun**:

1. Click **Clear All Outputs**
2. Click **Restart**
3. Click **Run All**

Then scroll through the notebook. Every code cell should have an execution count, and the numbers should go up in order from top to bottom. Do a clean rerun any time the notebook behaves strangely (for example, output that does not match the code above it).

If Python cannot carry out a line of code, it stops and shows an **error** message below the cell. Run All stops at the first cell with an error, and the cells below it are not run. Fix the error, then do a clean rerun again.

### 4. Saving your work to GitHub

Your notebook exists in two places. The copy on your computer is where you work. The copy on **GitHub**, a website that stores code projects online, is the copy other people see, including anyone reviewing your work. Changes you make on your computer don't reach GitHub until you send them there.

A few terms:

* **Repository** (or **repo**): a project folder whose history of changes is tracked. Your repository has a **local** copy on your computer and a **remote** copy on GitHub.
* **Git**: the tool that tracks those changes. VS Code provides buttons for Git in its **Source Control** view.
* **Save**: writes your changes to the notebook file on your computer. Nothing is sent to GitHub.
* **Stage**: choose which saved changes to include in your next commit.
* **Commit**: record the staged changes as a snapshot in your local repository's history, along with a short **commit message** describing what changed.
* **Push**: upload your commits to the remote repository on GitHub. In VS Code, the **Sync Changes** button pushes your commits (and also downloads any new commits from GitHub).

To save your work to GitHub:

1. **Save** the notebook: press **Ctrl + S**, or choose **File > Save**. While a file has unsaved changes, its tab at the top of VS Code shows a dot (●) instead of an ×.
2. **Open Source Control**: click the Source Control icon in the Activity Bar (the column of icons on the far left of VS Code), or press **Ctrl + Shift + G**. Your saved notebook appears under **Changes**.
3. **Stage**: hover over the file name and click the **+** (Stage Changes). The file moves to **Staged Changes**.
4. **Commit**: type a short commit message in the message box, then click **Commit**.
5. **Sync (push)**: click **Sync Changes**. If VS Code asks you to confirm, click **OK**.
6. **Confirm**: open your repository page on GitHub in a web browser and refresh it. Your commit message appears next to the notebook's file name. Click the file to see the notebook, including its output.
