### 1. Variables and printing

A **variable** is a name that stores a value so you can use it later in your code. You create a variable with `=`, the **assignment operator**. In Python, `=` does not mean "equals" the way it does in math. It means "store the value on the right in the name on the left." Read `station_name = "Elk 1"` as "station_name *gets* `"Elk 1"`."

Text written inside quotes is called a **string**. Numbers are written without quotes.

A line starting with `#` is a **comment**: a note for people reading the code. Python ignores it.

```python
# Store the string "Elk 1" in a variable named station_name
station_name = "Elk 1"

# Store the number 5320 in a variable named elevation_ft
elevation_ft = 5320
```

`print()` is a built-in **function**, a ready-made command that performs a task. Here, the task is displaying a value below the cell. Section 2 covers functions in more detail.

Quotes change what `print()` displays. Without quotes, Python treats the word as a variable name and prints the value stored in it. With quotes, Python treats it as a string and prints the text exactly as written.

```python
# We will use variable station_name from above: station_name = "Elk 1"

# No quotes: Python prints the value stored in the variable
print(station_name)

# Quotes: Python prints the text exactly as written
print("station_name")
```

`print()` can display several values at once when they are separated by commas. It prints them on one line with a space between each. Putting a string label before a variable shows which value is which, especially when a cell prints several variables.

```python
# We will use variable station_name from above: station_name = "Elk 1"
# We will use variable elevation_ft from above: elevation_ft = 5320

# A string label, a comma, then the variable
print("station_name:", station_name)
print("elevation_ft:", elevation_ft)
```

Variable names can contain letters, numbers, and underscores, but cannot start with a number or contain spaces or other special characters. Names are **case-sensitive**: `station_name` and `Station_Name` are two different variables. The Python convention is to write variable names as lowercase words joined by underscores.

```python
# Legal variable names: lowercase words joined by underscores
checker_name = "B Stewart"
station_2_name = "Bear Butte 1"

print("checker_name:", checker_name)
print("station_2_name:", station_2_name)

# Illegal variable names -- commented out because they cause errors
# 2nd_station = "Elk 2"       # cannot start with a number
# check-date = "2021-12-21"   # - is the subtraction sign
# station name = "Elk 2"      # spaces are not allowed
# qual$ = "Fair"              # special characters are not allowed
```

Assigning a new value to an existing variable replaces the old value. This is called **reassignment**.

```python
station_qual = "Fair"
print("station_qual:", station_qual)

# Reassignment replaces the old value
station_qual = "Excellent"
print("station_qual:", station_qual)
```

Once a cell creating a variable has been run, any cell you run afterward can use that variable. What matters is the order cells are *run*, not the order they appear on the page. If a cell uses a variable whose cell hasn't been run yet, Python stops with an error called a **NameError**.

```python
# We will use variable station_name from above: station_name = "Elk 1"
# We will use variable checker_name from above: checker_name = "B Stewart"
print(station_name, "was checked by", checker_name)
```

Errors are a normal part of writing code. When Python can't run a line, it stops and shows an **error message** below the cell. Read the last line first: it names the type of error and briefly describes the problem. The message also points to the line where Python got stuck. For a first-time coder, the most common causes are a missing quote, a missing parenthesis, or a misspelled variable name.

### 2. Data types and the `type()` function

Every value has a **data type**, which determines what kind of value it is and what you can do with it. The four basic data types are:

* `int` (integer): a whole number, positive or negative, with no decimal point
* `float` (floating-point number): a number with a decimal point
* `str` (string): text, written inside quotes
* `bool` (boolean): either `True` or `False`, always capitalized

Lists and dictionaries are also data types; they are covered separately.

`type()` is a function that tells you a value's data type. You **call** a function by writing its name followed by parentheses. The value inside the parentheses is the function's **argument**, and the value the function hands back is its **return value**. A return value can be stored in a variable, like any other value. `type()` reports a data type in the form `<class 'int'>`; for now, read that simply as `int`.

```python
# int: a whole number
w_id = 1

# Call type() with w_id as its argument, and store the return value
w_id_type = type(w_id)

print("w_id:", w_id)
print("w_id type:", w_id_type)
```

A function call can also go directly inside `print()`. Python runs `type()` first, then `print()` displays its return value. The rest of this notebook uses this shorter form.

```python
# float: a number with a decimal point
x_coord = 610124.0815
print("x_coord:", x_coord)
print("x_coord type:", type(x_coord))

# str: text inside quotes
# We will use variable station_name from Section 1: station_name = "Elk 1"
print("station_name:", station_name)
print("station_name type:", type(station_name))

# bool: True or False
is_checked = True
print("is_checked:", is_checked)
print("is_checked type:", type(is_checked))
```

Quotes decide the data type. A number written inside quotes is a string, not a number, even though both print the same way. The data type controls how a value behaves: `* 2` doubles a number, but repeats a string twice.

```python
# We will use variable elevation_ft from Section 1: elevation_ft = 5320
elevation_text = "5320"  # the same digits, inside quotes

# Both print the same...
print("elevation_ft:", elevation_ft)
print("elevation_text:", elevation_text)

# ...but they are different data types...
print("elevation_ft type:", type(elevation_ft))
print("elevation_text type:", type(elevation_text))

# ...so they behave differently
print("elevation_ft * 2:", elevation_ft * 2)
print("elevation_text * 2:", elevation_text * 2)
```

### 3. Math with numbers

Python can do math with `int` and `float` values using **arithmetic operators**: `+` (add), `-` (subtract), `*` (multiply), and `/` (divide). The data type of the result follows two rules:

* If any value in the calculation is a `float`, the result is a `float`
* Division with `/` always returns a `float`, even when the result is a whole number

```python
# We will use variable elevation_ft from Section 1: elevation_ft = 5320

# int + int returns an int
raised_ft = elevation_ft + 100
print("raised_ft:", raised_ft)
print("raised_ft type:", type(raised_ft))

# int * float returns a float, because one of the values is a float
elevation_m = elevation_ft * 0.3048  # 1 foot = 0.3048 meters
print("elevation_m:", elevation_m)
print("elevation_m type:", type(elevation_m))

# / always returns a float, even when the result is a whole number
half_ft = elevation_ft / 2
print("half_ft:", half_ft)
print("half_ft type:", type(half_ft))
```

Python follows the usual order of operations: `*` and `/` are calculated before `+` and `-`. Parentheses `()` control which part is calculated first.

```python
elk_1_elev_ft = 5320
elk_2_elev_ft = 5480

# Without parentheses, / runs first: 5320 + (5480 / 2)
avg_wrong_ft = elk_1_elev_ft + elk_2_elev_ft / 2
print("Without parentheses:", avg_wrong_ft)

# With parentheses, + runs first: (5320 + 5480) / 2
avg_elev_ft = (elk_1_elev_ft + elk_2_elev_ft) / 2
print("With parentheses:", avg_elev_ft)
```

### 4. Converting between data types

Values don't always arrive as the data type you need. For example, every value read from a CSV or text file arrives as a string, even a coordinate, and you can't do math with a string. **Converting** a value turns it into a different data type. The functions `int()`, `float()`, and `str()` each take one argument and return it converted to that data type.

```python
# A coordinate read from a CSV file arrives as a string
x_text = "610124.0815"
print("x_text:", x_text)
print("x_text type:", type(x_text))

# float() converts a string that looks like a number into a float
x_value = float(x_text)
print("x_value:", x_value)
print("x_value type:", type(x_value))

# Now math works: convert meters to kilometers
x_km = x_value / 1000
print("x_km:", x_km)
```

`int()` converts a `float` to an `int` by removing everything after the decimal point. It does **not** round. To round to the nearest whole number, use the `round()` function instead.

```python
# We will use variable elevation_m from Section 3: elevation_m = 1621.536

# int() removes the decimal part
elevation_m_int = int(elevation_m)
print("int(elevation_m):", elevation_m_int)

# round() rounds to the nearest whole number
elevation_m_round = round(elevation_m)
print("round(elevation_m):", elevation_m_round)
```

`int()` also converts a string of digits into an `int`, and `str()` converts any value into a string.

```python
# int() converts a string of digits into an int
year_text = "2021"
check_year = int(year_text)
print("check_year:", check_year)
print("check_year type:", type(check_year))

# str() converts a number into a string
# We will use variable w_id from Section 2: w_id = 1
w_id_text = str(w_id)
print("w_id_text:", w_id_text)
print("w_id_text type:", type(w_id_text))
```
