### 2. Finding the data

To extract the zip file, Python first needs its path. In the Strings notebook, you learned that a file path is stored as a string. The `os` module includes a smaller module, `os.path`, with functions for building and checking paths.

The **current working directory** is the folder Python is running in. `os.getcwd()` returns its path as a string. The path to the zip file starts here.

```python
# getcwd() returns the path of the current working directory
current_folder = os.getcwd()
print(f"current_folder: {current_folder}")
```

Check the printed path. It should end with the `code-alongs` folder that contains this notebook. The zip file is in the `data` folder at the top level of your repository, two folders above `code-alongs`:

    your repository folder
    ├── data
    │   └── austin_data.zip
    └── M2-python-language-fundamentals
        └── code-alongs
            └── course-data-setup.ipynb

A **relative path** starts from the current working directory instead of from the drive. In a relative path, `..` means the **parent folder**, the folder one level up, so `..\..` goes up two folders.

An **absolute path** is a full path that starts at the drive, such as `C:\`. `os.path.abspath()` turns a relative path into an absolute path.

```python
# Relative path from code-alongs up two folders to the repository
# folder, then down into data. Windows paths use a raw string
zip_relative = r"..\..\data\austin_data.zip"
print(f"zip_relative: {zip_relative}")

# Convert the relative path to an absolute path
zip_path = os.path.abspath(zip_relative)
print(f"zip_path: {zip_path}")
```

Before extracting the zip file, confirm that it is where the path says it is. `os.path.exists()` returns `True` if a file or folder exists at a path, and `False` if it does not. If it returns `False`, check the printed `zip_path` for a misspelled name or a wrong number of `..`.

```python
# exists() returns True if a file or folder exists at the path, and False
# if it does not
zip_exists = os.path.exists(zip_path)
print(f"{zip_path} exists? {zip_exists}")
```

### 3. Choosing where the data goes

The data goes in a working folder outside your repository. The recommended location is `C:\workspace\gist`, and the rest of this notebook uses it. If you choose a different location, keep it outside your repository and follow the naming rules below.

⚠️ **Important**: Every folder name along the path to your data and outputs should contain only letters, numbers, and underscores. Avoid:

* spaces, such as `GIS Data`
* special characters, such as `&`, `#`, `%`, `@`, `'`, `(`, `)`, and `-`
* folders whose names you do not control, such as a OneDrive folder named `OneDrive - University of Arizona`

Some geoprocessing tools raise an error when a path contains spaces, and Esri recommends avoiding spaces and special characters in folder and file names.

```python
# Store the path of the working folder as a raw string
# Update this path if you choose a different folder
working_folder = r"C:\workspace\gist"
print(f"working_folder: {working_folder}")
```

`os.makedirs()` creates a folder. If a folder on the way to it does not exist yet, such as `C:\workspace`, it creates that folder too. By default, `os.makedirs()` raises an error if the folder already exists. Its optional parameter `exist_ok` has a default value of `False`. Passing the keyword argument `exist_ok=True` tells it to do nothing when the folder already exists, so the cell can be rerun safely. You may see `os.mkdir()` in other code. It creates one folder only, and raises an error if a folder on the way to it is missing.

```python
# Create the working folder, and C:\workspace if it does not exist yet
os.makedirs(working_folder, exist_ok=True)
```

`os.path.isdir()` returns `True` if a path exists and is a folder, and `False` otherwise.

```python
# Confirm that the working folder exists and is a folder
print(f"{working_folder} is a folder? {os.path.isdir(working_folder)}")
```

### 4. Extracting the data

To **extract** a zip file is to copy the files and folders stored in it out to a folder, where they can be used. `zipfile.ZipFile()` opens a zip file so Python can read it.

A zip file that Python opens should also be closed when Python is done with it. A `with` statement does this for you: it opens the zip file, gives it the name after `as`, runs the indented code block, and closes the zip file when the code block ends. Inside the code block, the `.extractall()` method extracts every file and folder in the zip file to the folder you pass to it.

```python
# We will use variables zip_path from Section 2 and working_folder from
# Section 3
print(f"zip_path: {zip_path}")
print(f"working_folder: {working_folder}")

# Open the zip file, extract everything into the working folder, and
# close the zip file when the code block ends
with zipfile.ZipFile(zip_path) as austin_zip:
    print("Extracting austin_data.zip")
    austin_zip.extractall(working_folder)

print("Extraction finished")
```

The zip file stores everything inside a folder named `austin_data`, so the data is now in the `austin_data` folder inside your working folder. If you run the cell again, `.extractall()` replaces each extracted file with the copy from the zip file without raising an error.

### 5. Verifying the data

The extraction finished without an error, but that alone does not confirm the data is complete. We can now check the contents of the austin_data folder.

So far, you have typed each path in full. When a path builds on one you already have, such as a folder inside the working folder, you can join the parts instead. `os.path.join()` joins folder and file names into one path and places the correct separator between them, `\` on Windows, so you do not type any backslashes yourself.

```python
# The data is in the austin_data folder inside the working folder
data_folder = os.path.join(working_folder, "austin_data")
print(f"data_folder: {data_folder}")
```

`os.listdir()` returns a list of the names of every file and folder in a folder. Python does not guarantee any particular order for the names.

```python
# listdir() returns the names of the files and folders in data_folder
data_contents = os.listdir(data_folder)
print(f"data_contents: {data_contents}")
print(f"Number of items: {len(data_contents)}")
```

`os.listdir()` looks inside one folder only. `os.walk()` visits a folder and every folder inside it, one folder at a time. For each folder, it gives back three values as a tuple, the read-only list type from the Lists notebook: the folder's path, a list of its subfolder names, and a list of its file names. A `for` loop can store the three values in three variables, the same way a loop over a dictionary's `.items()` stores each key and value in two variables.

```python
# For each folder, print its path, its subfolders, its number of files,
# and the names of up to 5 of its files
for folder_path, subfolders, files in os.walk(data_folder):
    print(f"Folder: {folder_path}")
    print(f"\tSubfolders: {subfolders}")
    print(f"\tNumber of files: {len(files)}")
    print(f"\tFirst 5 files: {files[:5]}")
```

⚠️ **Important**: Look at the output for `Austin.gdb`. To the `os` module, a file geodatabase is an ordinary folder full of internal files with names such as `a00000001.gdbtable`, not the feature classes and tables you see in ArcGIS Pro. The `os` module does not understand GIS data. ArcPy provides list functions that do, and you will use them in the ArcPy notebooks.
