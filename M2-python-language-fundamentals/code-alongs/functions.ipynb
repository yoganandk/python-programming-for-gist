### 1. Defining, calling, and returning

You have been using built-in functions like `len()` that give back a value, such as the number of items in a list. You can write your own functions that work the same way.
A **function** is a named block of code that runs only when you ask it to. To **define** a function, write `def`, the function's name, parentheses `()`, and a colon `:`. Every line that belongs to the function's code block is indented underneath, the same way as in an `if` statement or a loop.
Inside the code block, `return` sends a value back to the place where the function was used. This value is called the **return value**.

```python
# Define a function named get_report_header
# Running this cell only defines the function. Its code block does not
# run yet, so nothing is printed
def get_report_header():
    header = "Pour Point Station Report"
    return header  # Send the value stored in header back
```

To **call** a function, write its name followed by parentheses, the same way you call `print()` or `len()`. Calling a function runs its code block. You can store the return value in a variable.

```python
# We will use function get_report_header from the previous cell
# Call the function and store its return value in a variable
report_header = get_report_header()
print(f"report_header: {report_header}")

# A function can be called as many times as you need
second_header = get_report_header()
print(f"second_header: {second_header}")
```

A function's code block can contain anything you have learned so far, including lists, loops, and conditional statements. A function can return any type of value, such as a list.

```python
# Define a function that returns a list of the Bear Butte stations
def get_bear_butte_stations():
    # List stores station names
    station_names = ["Elk 1", "Bear Butte 1", "Bear Butte 2",
                     "Bear Butte 3", "Elk 2", "Whitewood 1"]

    # Empty list will store only the Bear Butte stations
    bear_butte_stations = []
    for name in station_names:
        if name.startswith("Bear Butte"):
            bear_butte_stations.append(name)

    return bear_butte_stations  # Send the list back


# Call the function and store the returned list in a variable
bear_butte_list = get_bear_butte_stations()
print(f"bear_butte_list: {bear_butte_list}")

# The returned list can be used like any other list
print(f"Number of Bear Butte stations: {len(bear_butte_list)}")
```

### 2. Parameters and arguments

The functions in Section 1 return the same value every time they are called. Built-in functions like `len()` are more useful because you give them a value to work with: `len(station_names)` counts the items in whichever list you pass in.
A **parameter** is a variable name listed inside the parentheses when you define a function. An **argument** is the value you pass in for that parameter when you call the function. Inside the code block, the parameter stores the argument's value. Multiple parameters, and multiple arguments, are separated by commas.

```python
# name and qual are parameters
def describe_station(name, qual):
    description = f"{name} has a quality rating of {qual}"
    return description


# "Elk 1" and "Fair" are arguments
elk_description = describe_station("Elk 1", "Fair")
print(f"elk_description: {elk_description}")

# Call the same function with different arguments
bear_description = describe_station("Bear Butte 1", "Excellent")
print(f"bear_description: {bear_description}")
```

Arguments are matched to parameters by position: the first argument goes to the first parameter, the second argument to the second parameter, and so on. Python does not check whether the values make sense, so arguments in the wrong order produce a wrong result without an error, a logical error. Calling a function with too few arguments raises a `TypeError`.

```python
# We will use function describe_station from the previous cell
# Arguments in the wrong order: "Fair" goes to name, "Elk 1" goes to qual
swapped_description = describe_station("Fair", "Elk 1")
print(f"swapped_description: {swapped_description}")

# Too few arguments raises an error
# describe_station("Elk 1")
# TypeError: describe_station() missing 1 required positional argument: 'qual'
```

A function can contain an `if`/`elif`/`else` statement with a `return` in each branch. `return` ends the function immediately, so only the value from the first branch whose condition is `True` is sent back. Calling the function inside a loop classifies every value in a list.

```python
# Define a function that returns an elevation class
def get_elevation_class(elevation_ft):
    # elif only runs when the earlier conditions were False, so each
    # branch only needs to check one boundary
    if elevation_ft >= 9000:
        return "above 9000ft"
    elif elevation_ft >= 6000:
        return "between 6000ft and 9000ft"
    elif elevation_ft >= 3000:
        return "between 3000ft and 6000ft"
    else:
        return "below 3000ft"


# List stores elevation values in feet
elevations_ft = [9500, 6500, 4200, 2800]
print(f"elevations_ft: {elevations_ft}")

# Call the function once for each elevation in the list
for elevation in elevations_ft:
    elevation_class = get_elevation_class(elevation)
    print(f"\tElevation ({elevation}ft) is {elevation_class}")
```

A parameter can store any type of value, including a dictionary. The function below takes one station record and returns a one-line summary.

```python
# Define a function that returns a summary of one station record
def summarize_station(station):
    summary = (f"{station['NAME']} (W_ID {station['W_ID']}): "
               f"{station['QUAL']}, checked by {station['CHK_BY']} "
               f"on {station['CHK_DATE']}")
    return summary


# Dictionaries store two station records
bear_butte_1 = {
    "W_ID": 2,
    "NAME": "Bear Butte 1",
    "QUAL": "Excellent",
    "CHK_BY": "A Thomas",
    "CHK_DATE": "2022-12-21"
}
whitewood_1 = {
    "W_ID": 6,
    "NAME": "Whitewood 1",
    "QUAL": "Very Good",
    "CHK_BY": "A Wilson",
    "CHK_DATE": "2023-01-18"
}

# Call the function with each dictionary as the argument
bear_summary = summarize_station(bear_butte_1)
print(f"bear_summary: {bear_summary}")

whitewood_summary = summarize_station(whitewood_1)
print(f"whitewood_summary: {whitewood_summary}")
```

### 3. Local scope and functions that only print

A variable created inside a function's code block, including its parameters, exists only while the function runs. This is called **local scope**. Once the function finishes, its local variables are gone, and the return value is the only thing that comes back out.

```python
# Define a function that creates a local variable
def get_station_label(name, qual):
    label = f"{name} [{qual}]"  # label is a local variable
    return label


poor_label = get_station_label("Bear Butte 4", "Poor")
print(f"poor_label: {poor_label}")

# label only existed while the function ran, so this raises an error
# print(label)
# NameError: name 'label' is not defined
```

A function does not need a `return`. A function without one still gives back a value when it is called: `None`, the same value `.get()` returns for a missing key. This is a common source of confusion. A function that prints a result shows it on screen, but gives back nothing you can store and use later in your code.

```python
# This function prints the description instead of returning it
def print_description(name, qual):
    print(f"\t{name} has a quality rating of {qual}")


# Calling the function prints the description
printed_result = print_description("Elk 2", "Excellent")

# No value was returned, so the variable stores None
print(f"printed_result: {printed_result}")

# We will use function describe_station from Section 2, which returns
# the description instead of printing it
returned_result = describe_station("Elk 2", "Excellent")
print(f"returned_result: {returned_result}")
```

### 4. Default parameter values

You have already used optional values with built-in functions and methods, such as `.get("ELEV", "No elevation recorded")`, `enumerate(buffer_radii_m, start=1)`, and `.sort(reverse=True)`. Your own functions can have optional parameters too.
A **default** is a value given to a parameter in the function definition, using `=`. If a call does not pass an argument for that parameter, the parameter stores the default.
A **keyword argument** is an argument passed using the parameter's name, like `start=1`. It makes clear which parameter the value is for.

```python
# step_m has a default value of 5000
def get_buffer_radii(max_m, step_m=5000):
    # Empty list will store each buffer radius, in meters
    radii = []
    radius = step_m
    while radius <= max_m:
        radii.append(radius)
        radius += step_m
    return radii


# No argument for step_m, so step_m stores the default 5000
default_radii = get_buffer_radii(20000)
print(f"default_radii: {default_radii}")

# A second argument replaces the default
small_step_radii = get_buffer_radii(20000, 2500)
print(f"small_step_radii: {small_step_radii}")

# A keyword argument names the parameter the value is for
large_step_radii = get_buffer_radii(20000, step_m=10000)
print(f"large_step_radii: {large_step_radii}")
```

Parameters with a default must come after all parameters without one. Otherwise Python raises a `SyntaxError` when the function is defined.
