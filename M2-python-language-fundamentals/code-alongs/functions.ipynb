### 1. Anatomy of a function

You have been using built-in functions like `len()` that give back a value, such as the number of items in a list. You can write your own functions that work the same way.

A **function** is a named block of code that runs only when you ask it to. To **define** a function, you write a function definition, which has these parts:

* `def`, a **keyword** (a word Python reserves for its own use) that starts the definition
* the function's name
* parentheses `()`, which hold the function's **parameters**, variable names that store the values passed into the function. The parentheses are empty if the function has no parameters
* a colon `:`
* an optional **docstring**, a string in triple quotes `"""` on the first line of the code block that describes what the function does
* the code block, every line indented underneath the definition line, the same way as in an `if` statement or a loop

Inside the code block, `return` sends a value back to the place where the function was used. This value is called the **return value**. A return value can be any type of value, such as a number, a string, a list, a dictionary, or even `None`.

```python
# Define a function that counts the items in a list
def count_items(items):
    """Return the number of items in a list.

    Args:
        items: A list of values to count.

    Returns:
        The number of items in the list, as an integer.
    """
    # The indented lines below the def line are the function's code block
    count = 0
    for item in items:
        count += 1
    return count  # return sends the value in count back
```

A docstring can be a single line, or several lines for a function that needs more explanation. The docstring in `count_items` has three parts:

* a one-line summary of what the function does, followed by a blank line
* `Args:`, short for arguments, which lists each parameter by name, followed by what it should store
* `Returns:`, which describes the value the function sends back with `return`

The closing `"""` goes on its own line. The `Args:` and `Returns:` headings are a common convention, not a rule Python enforces, and you will see them often in existing code and AI-generated code. Short functions often use a one-line docstring instead.

Unlike a `#` comment, Python keeps a docstring with the function, so `help()` can display it. The next cell shows this.

```python
# help() displays a function's name, its parameters, and its docstring
help(count_items)
```

To **call** a function, write its name followed by parentheses, the same way you call `print()` or `len()`. 

Calling a function runs its code block. The value you put inside the parentheses is called an **argument**. While the function runs, the parameter stores the argument's value: in the next cell, the parameter `items` stores the list in `station_names`. You can store the return value in a variable.

```python
# List stores station names
station_names = ["Elk 1", "Bear Butte 1", "Bear Butte 2", "Whitewood 1"]
print(f"station_names: {station_names}")

# Call count_items with station_names as the argument, and store the
# return value in a variable
station_count = count_items(station_names)
print(f"station_count: {station_count}")

# count_items() gives the same result as the built-in len()
print(f"len(station_names): {len(station_names)}")
```

### 2. Positional and keyword arguments

The function in Section 1 has one parameter. A function can have several, separated by commas in the definition. When you call it, the arguments are separated by commas the same way. You have already used a built-in function like this: `round(elevation_m, 2)` passes two arguments, the value to round and the number of decimal places to keep.

An argument can be a variable, like in Section 1, or a value typed directly into the call.

```python
# Define a function with two parameters, distance and unit
def create_linear_unit(distance, unit):
    """Return a linear unit string, such as "5000 Meters"."""
    linear_unit = f"{distance} {unit}"
    return linear_unit
```

A **positional argument** is matched to a parameter by its position: the first argument goes to the first parameter, the second argument to the second parameter, and so on.

```python
# Example 1: Positional Arguments

# Positional arguments: 5000 goes to distance, "Meters" goes to unit
buffer_distance = create_linear_unit(5000, "Meters")
print(f"buffer_distance: {buffer_distance}")

# Positional arguments in the wrong order: "Meters" goes to distance and
# 5000 goes to unit. No error is raised, but the result is wrong
swapped_distance = create_linear_unit("Meters", 5000)
print(f"swapped_distance: {swapped_distance}")

# Too few arguments raises an error
# create_linear_unit(5000)
# TypeError: create_linear_unit() missing 1 required positional
# argument: 'unit'
```

A **keyword argument** is matched to a parameter by name, written as `parameter_name=value`. Because each value is labeled, keyword arguments can be passed in any order. You can combine both kinds in one call, but every positional argument must come before the first keyword argument.

```python
# Example 2: Keyword Arguments

# Keyword arguments in the same order as the parameters
keyword_distance = create_linear_unit(distance=5000, unit="Meters")
print(f"keyword_distance: {keyword_distance}")

# Keyword arguments in a different order give the same result
reordered_distance = create_linear_unit(unit="Meters", distance=5000)
print(f"reordered_distance: {reordered_distance}")

# A positional argument first, then a keyword argument
mixed_distance = create_linear_unit(5000, unit="Meters")
print(f"mixed_distance: {mixed_distance}")

# A positional argument after a keyword argument raises an error
# create_linear_unit(distance=5000, "Meters")
# SyntaxError: positional argument follows keyword argument
```

A variable created inside a function's code block, including its parameters, exists only while the function runs. This is called **local scope**. Once the function finishes, its local variables are gone, and the return value is the only thing that comes back out.

```python
# Inside the function, linear_unit is a local variable
short_distance = create_linear_unit(250, "Meters")
print(f"short_distance: {short_distance}")

# linear_unit only existed while the function ran, so this raises an
# error
# print(linear_unit)
# NameError: name 'linear_unit' is not defined
```

### 3. Return values

The functions in Sections 1 and 2 use `return` once, on the last line of the code block. A function can also contain an `if`/`elif`/`else` statement with a `return` in each branch. `return` ends the function immediately, so only the value from the first branch whose condition is `True` is sent back.

```python
# Define a function that returns an elevation class
def get_elevation_class(elevation_ft):
    """Return the elevation class for an elevation in feet."""
    # elif only runs when the earlier conditions were False, so each
    # branch only needs to check one boundary
    if elevation_ft >= 9000:
        return "above 9000ft"
    elif elevation_ft >= 6000:
        return "between 6000ft and 9000ft"
    elif elevation_ft >= 3000:
        return "between 3000ft and 6000ft"
    else:
        return "below 3000ft"


# List stores elevation values in feet
elevations_ft = [9500, 6500, 4200, 2800]
print(f"elevations_ft: {elevations_ft}")

# Call the function once for each elevation in the list
for elevation in elevations_ft:
    elevation_class = get_elevation_class(elevation)
    print(f"\tElevation {elevation}ft is {elevation_class}")
```

A function does not need a `return`. A function without one still gives back a value when it is called: `None`. This is a common source of confusion. A function that prints a result shows it on screen, but gives back nothing you can store and use later in your code.

```python
# This function prints the linear unit instead of returning it
def print_linear_unit(distance, unit):
    """Print a linear unit string, such as "5000 Meters"."""
    print(f"\t{distance} {unit}")


# Calling the function prints the linear unit
printed_result = print_linear_unit(1000, "Meters")

# No value was returned, so the variable stores None
print(f"printed_result: {printed_result}")

# We will use function create_linear_unit from Section 2, which returns
# the linear unit instead of printing it
returned_result = create_linear_unit(1000, "Meters")
print(f"returned_result: {returned_result}")
```

### 4. Default parameter values

A parameter is **optional** when it has a **default**, a value given to the parameter in the function definition using `=`. If a call does not pass an argument for an optional parameter, the parameter stores its default. `help()` shows each optional parameter with its default value.

You have already used a built-in function with an optional parameter: `round(elevation_m)` works with one argument because the number of decimal places is optional.

```python
# Define create_linear_unit again, this time with a default value for
# unit. Defining a function with the same name replaces the earlier one
def create_linear_unit(distance, unit="Meters"):
    """Return a linear unit string, such as "5000 Meters".

    Args:
        distance: The distance, as a number.
        unit: The unit of the distance. Defaults to "Meters".

    Returns:
        The distance and unit as one string.
    """
    linear_unit = f"{distance} {unit}"
    return linear_unit


# help() shows unit='Meters', so unit is optional
help(create_linear_unit)
```

```python
# No argument for unit, so unit stores the default "Meters"
default_unit = create_linear_unit(5000)
print(f"default_unit: {default_unit}")
```

```python
# A positional argument replaces the default
feet_unit = create_linear_unit(500, "Feet")
print(f"feet_unit: {feet_unit}")
```

```python
# A keyword argument also replaces the default
miles_unit = create_linear_unit(3, unit="Miles")
print(f"miles_unit: {miles_unit}")
```

Parameters with a default must come after all parameters without one. Otherwise Python raises a `SyntaxError` when the function is defined.

```python
# A parameter without a default cannot follow one with a default
# def create_linear_unit(unit="Meters", distance):
# SyntaxError: non-default argument follows default argument
```

### 5. Functions that loop and call other functions

A function can contain a loop, and it can call another function. The function below takes a whole list of elevations, loops over it inside its code block, calls `get_elevation_class` from Section 3 for each item, and returns a dictionary that pairs each elevation, as the key, with its elevation class, as the value. The loop is written once, inside the function, so any list of elevations can be classified with a single call.

Building a larger function from smaller ones that each do one task keeps each function short and reusable. You will see this pattern often in existing code and AI-generated code.

```python
# Define a function that classifies every elevation in a list
def get_elevation_classes(elevations_ft):
    """Return a dictionary of elevation classes for a list of elevations.

    Args:
        elevations_ft: A list of elevations in feet.

    Returns:
        A dictionary with each elevation as a key and its elevation
        class as the value.
    """
    elevation_classes = {}
    for elevation in elevations_ft:
        # Call get_elevation_class from Section 3 on each item, and store
        # the result as the value for the elevation key
        elevation_classes[elevation] = get_elevation_class(elevation)
    return elevation_classes
```

```python
# We will use variable elevations_ft from Section 3
print(f"elevations_ft: {elevations_ft}")
all_classes = get_elevation_classes(elevations_ft)
print(f"all_classes: {all_classes}")

# Loop over the returned dictionary to print each elevation and its class
for elevation, elevation_class in all_classes.items():
    print(f"\tElevation {elevation}ft is {elevation_class}")
```

```python
# The same function works on any list of elevations
new_classes = get_elevation_classes([7200, 1500, 4356])
print(f"new_classes: {new_classes}")
```
