### 1. Creating lists and accessing items

A **list** stores multiple values in a single variable. Items go inside square brackets `[]`, separated by commas. Each item has an **index**, its position in the list, starting at 0, so you can access a single item (`my_list[0]`).

```python
# A list stores multiple related values in a single variable
station_names = ["Elk 1", "Bear Butte 1", "Bear Butte 2", "Whitewood 1"]
print(f"station_names: {station_names}")

# Lists can mix data types too -- a single station record works well
# as a list: an integer ID, two strings, and two floats
elk_list = [1, "Elk 1", 610124.0815, 4905101.303, "Fair"]
print(f"elk_list: {elk_list}")

# len() counts how many items are in a list, regardless of their type
print(f"Number of items in station_names: {len(station_names)}")  # 4 items
print(f"Number of items in elk_list: {len(elk_list)}")            # 5 items

# Access first item at index 0 -> "Elk 1"
elk_station = station_names[0]
print(f"First station: {elk_station}")

# Access item at index 2 -> "Bear Butte 2"
bear_2_station = station_names[2]
print(f"Station at index 2: {bear_2_station}")

# Access last item at index -1 -> "Whitewood 1"
whitewood_station = station_names[-1]
print(f"Last station: {whitewood_station}")
```

### 2. Checking and counting contents

The `in` operator checks whether a value exists in a list and returns a boolean, `True` or `False`. The `.count()` method counts how many times a value appears in a list, and `.index()` returns the index of the first occurrence of a value.

```python
# List stores quality ratings
quality_ratings = ["Fair", "Excellent", "Fair", "Excellent", "Very Good"]
print(f"quality_ratings: {quality_ratings}")

# The `in` operator checks membership and returns a boolean
has_excellent = "Excellent" in quality_ratings
print(f"Is Excellent in quality_ratings? {has_excellent}")       # True

# Python is case-sensitive, so this returns False
has_excellent_lower = "excellent" in quality_ratings
print(f"Is excellent in quality_ratings? {has_excellent_lower}")

# .count() returns how many times a value appears in a list
fair_count = quality_ratings.count("Fair")
print(f"Count of Fair: {fair_count}")  # 2

# .index() returns the index of the first occurrence of a value
excellent_index = quality_ratings.index("Excellent")
print(f"Index of Excellent: {excellent_index}")  # 1
```

### 3. Adding and removing items

Lists are **mutable**, meaning you can change their contents after creation, unlike strings. `.append()` adds an item to the end of a list, `.insert()` adds an item at a specific index, and `.extend()` adds multiple items at once.

```python
station_names = ["Elk 1", "Bear Butte 1", "Bear Butte 2", "Whitewood 1"]
print(f"station_names: {station_names}")

# .append() adds an item to the end
station_names.append("Elk 1")
print(f"After append: {station_names}")

# .insert() adds an item at the specified index
station_names.insert(1, "Black Butte 2")
print(f"After insert: {station_names}")

# New list stores two station names
more_stations = ["Whitewood 2", "Bear Butte 4"]

# .extend() adds multiple items at once
station_names.extend(more_stations)
print(f"After extend: {station_names}")
```

To remove items, `del` removes the item at a given index, `.pop()` removes an item at a given index (the last item by default) and returns it, and `.remove()` deletes the first item that matches a given value.

```python
# del removes the item at a given index
del station_names[-3]  # Remove third item from the end
print(f"After del: {station_names}")

# .pop() removes an item at the specified index and returns it
removed_station = station_names.pop(5)  # Remove item at index 5
print(f"Removed: {removed_station}")
print(f"After pop: {station_names}")

# .remove() removes the first matching value from the list
station_names.remove("Black Butte 2")
print(f"After remove: {station_names}")
```

### 4. Sorting and reversing

`.sort()` sorts a list in place, ascending by default, or descending with `reverse=True`. `.reverse()` reverses the current order of a list without sorting it.

```python
station_names = ['Elk 1', 'Bear Butte 1', 'Bear Butte 2', 'Whitewood 1', 'Bear Butte 4']
print(f"station_names: {station_names}")

# Sort alphabetically, ascending
station_names.sort()
print(f"Sorted ascending: {station_names}")

# Sort alphabetically, descending
station_names.sort(reverse=True)  # sorts alphabetically, descending
print(f"Sorted descending: {station_names}")

# Reverse the current order of the items, without sorting
station_names.reverse()
print(f"Reversed: {station_names}")
```

### 5. Converting between strings and lists

The `.split()` string method breaks a string into a list, using the given character(s) as the separator. The `.join()` string method does the opposite by combining a list of strings into one string, using the string it is called on as the separator between items.

```python
# Windows path to a made-up file
ws_path = r"c:\workspace\gist\data\pourpoints.csv"
print(f"ws_path: {ws_path}")

# .split() breaks a string into a list on the given separator \
path_parts = ws_path.split("\\")  # "\\" represents one backslash
print(f"path_parts: {path_parts}")

# .join() combines a list of strings into one string, using the given
# / separator between items
rejoined_path = "/".join(path_parts)
print(f"rejoined_path: {rejoined_path}")

# Using .split() to split a station name on spaces
butte_station = "Bear Butte 1"
butte_name_parts = butte_station.split(" ")
print(f"butte_name_parts: {butte_name_parts}")

# Recreate the station name using underscores
underscored_name = "_".join(butte_name_parts)
print(f"underscored_name: {underscored_name}")
```

### 6. Tuples: read-only lists

A **tuple** is like a list, but read-only. You create one using parentheses `()` instead of square brackets. You can index and slice a tuple the same way you access a list, but once it's created you cannot add, remove, or change its items.

```python
# A tuple looks like a list, but uses parentheses instead of square
# brackets
week_tuple = ("MON", "TUE", "WED", "THU", "FRI", "SAT", "SUN")
print(f"week_tuple: {week_tuple}")

# Indexing and slicing work the same way they do for lists
print(f"First day: {week_tuple[0]}")
print(f"Last day: {week_tuple[-1]}")

# Tuples are read-only -- trying to change an item raises an error
# week_tuple[0] = "Monday"
# TypeError: 'tuple' object does not support item assignment
```
