# Conditional Statements

A GIS script often needs to branch: skip a step if a value falls outside an expected range, take one action if a field matches a value and a different one if it doesn't, react differently depending on whether something is present in a list. Conditional statements are how Python handles this branching of code.

🎯 What You'll Learn

* Write `if` and `if`/`else` statements to run code only when a condition is met
* Compare numbers and strings using comparison operators
* Chain multiple conditions in order with `elif`
* Check string values with `.startswith()` and `.endswith()`, and check whether an item is in a list using the `in` operator
* Combine conditions with logical operators: `and`, `or`, and `not`
* Recognize what Python considers `True` or `False` using `bool()`

### 1. Writing `if` and `if`/`else` statements

A **condition** is an expression that evaluates to a boolean, `True` or `False`. Code indented underneath `if`, its code block, only runs when the condition is `True`.

```python
station_qual = "Excellent"  # Variable stores a string
print(f"station_qual: {station_qual}")

# Check if the string in the variable is "Excellent"
if station_qual == "Excellent":  # == checks if both strings are equal
    # If condition is True, execute lines in this code block
    print(f"\tCondition is True: station_qual is {station_qual}")
```

Adding `else` gives you a second code block that runs when the condition is `False`.

```python
station_qual = "Poor"  # Changing the string
print(f"station_qual: {station_qual}")

# Check if the string in the variable is "Excellent"
if station_qual == "Excellent":
    # If condition is True, execute lines in this code block
    print(f"\tCondition is True: station_qual is {station_qual}")
else:
    # If condition is False, execute lines in this code block
    print(f"\tCondition is False: station_qual is {station_qual}")
```

 `if` does not require an `else`.

```python
station_qual = "Fair"  # Changing the string
print(f"station_qual: {station_qual}")

# Check if the string in the variable is "Poor"
if station_qual == "Poor":
    # If condition is True, execute lines in this code block
    print(f"\tCondition is True: station_qual is {station_qual}")

# This line always runs -- it is outside the if code block
print("Return to lines of code outside the if code block")
```

`pass` does nothing. It is a placeholder for a code block that does not contain any other code.

```python
station_qual = "Poor"  # Changing the string
print(f"station_qual: {station_qual}")

# Check if the string in the variable is "Poor"
if station_qual == "Poor":
    # If condition is True, execute lines in this code block
    pass  # TODO: add check logic later
else:
    # If condition is False, execute lines in this code block
    print(f"\tCondition is False: station_qual is {station_qual}")

# This line always runs -- it is outside the if/else code block  
print("Return to lines of code outside the if code block")
```

### 2. Comparing values

Comparison operators, like `==`, `!=`, `<`, `>`, `<=`, `>=` compare two values and return a boolean, `True` or `False`, the same way the `==` check in Section 1 did. They work on numbers as well as strings.  
`!=` comparison operator checks if two values are not equal

```python
no_of_cities = 9
print(f"no_of_cities: {no_of_cities}")

# != checks if two values are not equal
if no_of_cities != 10:
    print(f"\tCondition is True: no_of_cities ({no_of_cities}) does not equal 10")
else:
    print(f"\tCondition is False: no_of_cities ({no_of_cities}) equals 10")
```

`<` comparison operator checks if the left value is less than the right value. Similarly, `>` comparison operator checks if the left value is greater than the right value.

```python
no_of_cities = 8
print(f"no_of_cities: {no_of_cities}")

# < checks if the left value is less than the right value
if no_of_cities < 10:
    print(f"\tCondition is True: no_of_cities ({no_of_cities}) is less than 10")
else:
    print(f"\tCondition is False: no_of_cities ({no_of_cities}) is not less than 10")
```

`>=` comparison operator checks if the left value is greater than or equal to the right value. Similarly, `<=` comparison operator checks if the left value is less than or equal to the right value.

```python
no_of_cities = 10
print(f"no_of_cities: {no_of_cities}")

# >= checks if the left value is greater than or equal to the right value
if no_of_cities >= 10:
    print(f"\tCondition is True: no_of_cities ({no_of_cities}) is greater than or equal to 10")
else:
    print(f"\tCondition is False: no_of_cities ({no_of_cities}) is less than 10")
```

### 3. Chaining conditions with `elif`

`elif` lets you check several conditions in order after an initial `if`, the first one that is `True` runs, and the rest are skipped. This makes it possible to test more than two branches without nesting `if` statements inside each other.

```python
# Variable stores an elevation value in feet
# Change this value and rerun the cell to test each branch below
elevation_ft = 6500
print(f"elevation_ft: {elevation_ft}")

if elevation_ft >= 9000:
    print(f"\tElevation ({elevation_ft}ft) is above 9000ft")
elif elevation_ft >= 6000:
    print(f"\tElevation ({elevation_ft}ft) is between 6000ft and 9000ft")
elif elevation_ft >= 3000:
    print(f"\tElevation ({elevation_ft}ft) is between 3000ft and 6000ft")
else:
    print(f"\tElevation ({elevation_ft}ft) is below 3000ft")
```

### 4. Checking strings and list membership

`.startswith()` and `.endswith()` check the start or end of a string and return a boolean `True` or `False` based on a substring match.

```python
# Variable stores a station name
# Change this value and rerun the cell to test both branches
station_name = "Bear Butte 3"
print(f"station_name: {station_name}")

# .startswith() checks whether a string starts with a given substring
if station_name.startswith("Bear Butte"):
    print(f"\t{station_name} belongs to the Bear Butte network")
else:
    print(f"\t{station_name} does not belong to the Bear Butte network")
```

```python
# Variable stores a filename
# Change this value and rerun the cell to test both branches
survey_file = "pourpoints.shp"
print(f"survey_file: {survey_file}")

# .endswith() checks whether a string ends with a given substring
if survey_file.endswith(".shp"):
    print(f"\t{survey_file} is a shapefile")
else:
    print(f"\t{survey_file} is not a shapefile")
```

The `in` operator checks whether a value exists in a list and returns a boolean `True` or `False`, useful for deciding which branch of code to run based on whether something is already present.

```python
# List stores station names already checked this season
checked_stations = ["Elk 1", "Bear Butte 1", "Bear Butte 2", "Whitewood 1"]
print(f"checked_stations: {checked_stations}")

# Variable stores a station name to look up
# Change this value and rerun the cell to test both branches
station_name = "Elk 2"
print(f"station_name: {station_name}")

# The in operator checks whether a value exists in a list
if station_name in checked_stations:
    print(f"\t{station_name} has already been checked")
else:
    print(f"\t{station_name} still needs to be checked")
```

### 5. Combining conditions with logical operators

Logical operators include `and` and `or` that help combine multiple conditions into one and `not` that reverses the condition's boolean value.  
`and` requires every condition to be `True`

```python
# We will use variable no_of_cities from Section 2: no_of_cities = 10
# We will use variable elevation_ft from Section 3: elevation_ft = 6500
print(f"no_of_cities: {no_of_cities}, elevation_ft: {elevation_ft}")

# and operator requires every condition to be True
if no_of_cities >= 10 and elevation_ft >= 6000:
    print("\tCondition is True: both conditions are met")
else:
    print("\tCondition is False: not both conditions are met")
```

`or` requires only one condition to be `True`

```python
# We will use variable no_of_cities from Section 2: no_of_cities = 10
# We will use variable elevation_ft from Section 3: elevation_ft = 6500
print(f"no_of_cities: {no_of_cities}, elevation_ft: {elevation_ft}")

# or requires only one condition to be True
if no_of_cities < 5 or elevation_ft >= 6000:
    print("\tCondition is True: at least one condition is met")
else:
    print("\tCondition is False: neither condition is met")
```

`not` reverses a condition's boolean value. 

```python
# We will use variable elevation_ft from Section 3: elevation_ft = 6500
print(f"elevation_ft: {elevation_ft}")

# not reverses a condition's boolean value
if not elevation_ft >= 9000:
    print(f"\tElevation ({elevation_ft}ft) is not extreme")
else:
    print(f"\tElevation ({elevation_ft}ft) is extreme")
```

Python evaluates `and` before `or`, so combining them without parentheses may not group the way you intend. Using parentheses makes the grouping explicit.  

```python
# Without parentheses, Python evaluates and before or
# no_of_cities and elevation >= 9000 are checked first
if no_of_cities >= 20 and elevation_ft >= 9000 or elevation_ft >= 3000:
    print("\tWithout parentheses: True")
else:
    print("\tWithout parentheses: False")


# Parentheses change the order of the checks
# no_of_cities must be >= 20 AND at least one of the other two conditions
if no_of_cities >= 20 and (elevation_ft >= 9000 or elevation_ft >= 3000):
    print("\tWith parentheses: True")
else:
    print("\tWith parentheses: False")
```

### 6. Truthy and falsy values

Every value in Python has an inherent boolean value, even values that aren't explicitly `True` or `False`. Wrapping any value in `bool()` reveals whether Python treats it as **truthy** (equivalent to `True`) or **falsy** (equivalent to `False`). This is exactly what `if` is checking behind the scenes.  
`None`, `0`, an empty string `""`, and an empty list `[]` are all falsy; most other values are truthy.

```python
# Using bool()
print(f"bool(None): {bool(None)}")       # None is falsy
print(f"bool(0): {bool(0)}")             # 0 is falsy
print(f"bool(1): {bool(1)}")             # any nonzero number is truthy
print(f"bool(-1): {bool(-1)}")           # any nonzero number is truthy
print(f"bool(0.0): {bool(0.0)}")         # 0 is falsy
print(f"bool(''): {bool('')}")           # empty string is falsy
print(f"bool(' '): {bool(' ')}")         # whitespace string is truthy
print(f"bool('Fair'): {bool('Fair')}")   # non-empty string is truthy
print(f"bool([]): {bool([])}")           # empty list is falsy

station_names = ["Elk 1", "Bear Butte 1"]
print(f"bool(station_names): {bool(station_names)}")  # non-empty list
```

Because every value has a truthy or falsy meaning, `if` can check a variable directly instead of comparing it to something else.

```python
# Change this value and rerun the cell to test both branches
checked_by = ""
if checked_by:
    print(f"Checked by: {checked_by}")
else:
    print("Not yet checked")
```
