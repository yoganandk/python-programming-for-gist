### 1. Creating dictionaries and accessing values

A **dictionary** stores data as **key-value pairs**, also called **items**. The **key** is the name you use to look something up, and the **value** is the data stored under that name. Items go inside curly braces `{}`, with a colon `:` between each key and its value and commas between items. You access a value by using its key.

```python
# An empty dictionary uses curly braces with nothing inside
empty_dictionary = {}
print(f"empty_dictionary: {empty_dictionary}")

# The following dictionary stores one station record. Each key is
# a field name and each value is the attribute stored in that field
station = {
    "W_ID": 2,
    "NAME": "Bear Butte 1",
    "X_COORD": 611659.0625,
    "Y_COORD": 4913979.832,
    "QUAL": "Excellent",
    "ACTIVE": True
}
print(f"station: {station}")

# Access a value by putting its key inside square brackets
station_name = station["NAME"]
print(f"Station name: {station_name}")

# A key can also go directly inside an f-string. Use single quotes around
# the key when the f-string itself uses double quotes
print(f"Quality: {station['QUAL']}")
```

Keys do not have to be strings. Integer keys work well for looking something up by an ID number. Note that dictionaries do not have an index, so an integer key is a name for a value, not a position in the dictionary.

```python
# Integer keys and string values: look up a station name by its W_ID
station_lookup = {11: "Elk 1", 12: "Bear Butte 2", 13: "Bear Butte 3"}
print(f"station_lookup: {station_lookup}")

# 13 is the key here, not an index position
print(f"Station with W_ID 13: {station_lookup[13]}")
```

### 2. Adding and updating items

Dictionaries are **mutable**, so you can change their contents after they are created. Assigning a value to a key that does not exist yet adds a new item. Assigning a value to a key that already exists overwrites its value, because every key in a dictionary is unique. `len()` counts the items in a dictionary the same way it counts the items in a list.

```python
# We will use the dictionary stored in variable station from Section 1
print(f"station: {station}")
print(f"Number of items: {len(station)}")

# Assigning to a new key adds a new item
station["CHK_DATE"] = "2022-12-21"
print(f"After adding CHK_DATE: {station}")
print(f"Number of items: {len(station)}")

# Assigning to an existing key overwrites its value, no new item is added
station["QUAL"] = "Very Good"
print(f"After updating QUAL: {station}")
print(f"Number of items: {len(station)}")
```

You can also start with an empty dictionary and add items one at a time.

```python
# Start with an empty dictionary
new_station = {}
print(f"new_station: {new_station}")

# Add items one at a time
new_station["NAME"] = "Elk 2"
new_station["QUAL"] = "Excellent"
new_station["X_COORD"] = 614870.5044
new_station["Y_COORD"] = 4905514.835
new_station["ACTIVE"] = False
print(f"After adding five items: {new_station}")
```

### 3. Checking for keys with `in` and `.get()`

Accessing a key that does not exist raises an error called a `KeyError`. Keys in a dictionary are case-sensitive.

```python
# Accessing a key that does not exist raises an error
# print(station["ELEV"])
# KeyError: 'ELEV'

# Keys are case-sensitive, so this raises an error too
# print(station["qual"])
# KeyError: 'qual'
```

The `in` operator checks whether a key exists in a dictionary and returns a boolean, `True` or `False`, the same way it checks whether an item is in a list. With a dictionary, `in` checks the keys only, not the values.

```python
# We will use the dictionary stored in variable station from Section 1
print(f"station: {station}")

# in checks whether a key exists in the dictionary
has_qual = "QUAL" in station
print(f"Is QUAL a key in station? {has_qual}")                # True

has_elev = "ELEV" in station
print(f"Is ELEV a key in station? {has_elev}")                # False

# Keys are case-sensitive
has_lower_qual = "qual" in station
print(f"Is qual a key in station? {has_lower_qual}")          # False

# in checks keys only, not values
has_name_value = "Bear Butte 1" in station
print(f"Is Bear Butte 1 a key in station? {has_name_value}")  # False
```

Checking with `in` before accessing a key avoids a `KeyError`.

```python
# Variable stores a field name to look up
# Change this value and rerun the cell to test both branches
field = "ELEV"
print(f"field: {field}")

if field in station:
    # Key is in the dictionary
    print(f"\t{field}: {station[field]}")
else:
    # Key is not in the dictionary
    print(f"\t{field} is not a key in station")
```

The `.get()` method returns the value for a key without raising an error when the key is missing. It returns `None`, Python's value for "nothing here", unless you give it a second value to return instead, called a **default**.

```python
# .get() returns the value when the key exists
qual = station.get("QUAL")
print(f"QUAL: {qual}")

# .get() returns None when the key does not exist, instead of an error
elev = station.get("ELEV")
print(f"ELEV: {elev}")

# A second value sets the default to return when the key does not exist
elev_or_default = station.get("ELEV", "No elevation recorded")
print(f"ELEV: {elev_or_default}")
```

### 4. Removing items

`del` removes the item for a given key. `.pop()` removes the item for a given key and returns its value, so you can store it in a variable. Both raise a `KeyError` if the key does not exist.

```python
# We will use the dictionary stored in variable station from Section 1
print(f"station: {station}")

# del removes the item for a given key
del station["CHK_DATE"]
print(f"After del: {station}")

# .pop() removes the item for a given key and returns its value
removed_value = station.pop("ACTIVE")
print(f"Removed: {removed_value}")
print(f"After pop: {station}")
```

### 5. Looping over dictionaries

The `.keys()`, `.values()`, and `.items()` methods return all the keys, all the values, or all the key-value pairs in a dictionary. Each item from `.items()` is a tuple of `(key, value)`. The results look like lists wrapped in a label such as `dict_keys(...)`, and a `for` loop can iterate over them the same way it iterates over a list.

```python
# We will use the dictionary stored in variable station from Section 1
print(f"station: {station}")

# Print the keys
print(f"Keys: {station.keys()}")

# Print the values
print(f"Values: {station.values()}")

# Print the items (key-value pairs)
print(f"Items: {station.items()}")
```

Looping over `.keys()` gives you one key per pass through the loop. Use the key inside square brackets to get its value.

```python
# Print the keys
print(f"Keys: {station.keys()}")

# Loop over the keys
for key in station.keys():
    print(f"\t{key}: {station[key]}")
```

Looping over a dictionary directly also loops over its keys, so this produces the same output as the loop above. This is the shorter form you will see most often in existing code and AI-generated code.

```python
# Print the keys
print(f"Keys: {station.keys()}")

# Loop over the keys - alternative
for key in station:
    print(f"\t{key}: {station[key]}")
```

Looping over `.items()` gives you one `(key, value)` tuple per pass. Writing two loop variables after `for` separates the tuple into the key and the value.

```python
# Print the items (key-value pairs)
print(f"Items: {station.items()}")

# Loop over the items using two loop variables
for key, value in station.items():
    print(f"\t{key}: {value}")
```

### 6. Updating values inside a loop

A loop with an `if` inside can find specific items and update their values. Updating the value of an existing key inside a loop is fine. Adding or removing keys while looping over a dictionary raises a `RuntimeError`, so only update values inside the loop.

```python
# Example - Using loops and conditional statements to fix inconsistent
# capitalization in a dictionary

# Dictionary stores quality ratings for each station
# Two ratings have inconsistent capitalization
station_quals = {
    "Elk 1": "Fair",
    "Bear Butte 1": "excellent",
    "Bear Butte 2": "Fair",
    "Whitewood 1": "very good",
}
print(f"station_quals: {station_quals}")

# Loop on the keys
for name in station_quals:
    # Access the rating value using the name as key
    qual = station_quals[name]
    print(f"\tStation Name: {name} | Quality: {qual}")

    # Check whether the rating is not already in title case
    if qual != qual.title():
        # Update the value for this key
        station_quals[name] = qual.title()
        print(f"\t\tFixed {name}: {qual} -> {station_quals[name]}")

print(f"After fixing: {station_quals}")
```

### 7. Storing multiple values under one key

A value can also be a collection, like a list or a tuple. A tuple works well for values that belong together and should not change, such as an x/y coordinate pair.

```python
# A tuple stores the coordinate pair as a single value
elk_2 = {
    "NAME": "Elk 2",
    "QUAL": "Excellent",
    "XY": (614870.5044, 4905514.835),
}
print(f"elk_2: {elk_2}")

# Access the tuple using its key
xy = elk_2["XY"]
print(f"XY: {xy}")

# Then use an index to access each coordinate in the tuple
print(f"X: {xy[0]}")
print(f"Y: {xy[1]}")

# Alternatively, access the coordinates directly by chaining the key 
# and the index
print(f"X: {elk_2['XY'][0]}")
print(f"Y: {elk_2['XY'][1]}")
```
