# Dictionaries

A list stores values by position, so you have to remember that index 2 holds the x-coordinate and index 4 holds the quality rating. A dictionary stores values by name instead. A single station record becomes a set of field names, each paired with its value, much like one row of an attribute table. Dictionaries are how Python looks things up by name: a field value by its field name, a station name by its ID, a setting by its label.

* Check whether a key exists using `in` and `.get()`, and avoid a `KeyError`
* Remove items from a dictionary with `del` and `.pop()`
* Loop over a dictionary's keys, values, and items

### 1. Creating dictionaries and accessing values

A **dictionary** stores data as **key-value pairs**, also called **items**. The **key** is the name you use to look something up, and the **value** is the data stored under that name. Items go inside curly braces `{}`, with a colon `:` between each key and its value and commas between items. Instead of an index, you access a value by putting its key inside square brackets (`my_dict["key"]`).

```python
# An empty dictionary uses curly braces with nothing inside
empty_dict = {}
print(f"empty_dict: {empty_dict}")

# A dictionary stores one station record. Each key is a field name and
# each value is the attribute stored in that field
station = {
    "W_ID": 2,
    "NAME": "Bear Butte 1",
    "X_COORD": 611659.0625,
    "Y_COORD": 4913979.832,
    "QUAL": "Excellent",
    "CHK_BY": "A Thomas",
}
print(f"station: {station}")

# Access a value by putting its key inside square brackets
station_name = station["NAME"]
print(f"Station name: {station_name}")

# A key can also go directly inside an f-string. Use single quotes around
# the key when the f-string itself uses double quotes
print(f"Quality: {station['QUAL']}")
```

Keys do not have to be strings. Integer keys work well for looking something up by an ID number. Even though `station_lookup[3]` looks like indexing, `3` here is a key, not a position: dictionaries do not have indexes.

```python
# Integer keys and string values: look up a station name by its W_ID
station_lookup = {1: "Elk 1", 2: "Bear Butte 1", 3: "Bear Butte 2"}
print(f"station_lookup: {station_lookup}")

# 3 is the key here, not an index position
print(f"Station with W_ID 3: {station_lookup[3]}")
```

### 2. Adding and updating items

Dictionaries are **mutable**, like lists, so you can change their contents after they are created. Assigning a value to a key that does not exist yet adds a new item. Assigning a value to a key that already exists overwrites its value, because every key in a dictionary is unique. `len()` counts the items in a dictionary the same way it counts the items in a list.

```python
# We will use variable station from Section 1
print(f"station: {station}")
print(f"Number of items: {len(station)}")

# Assigning to a new key adds a new item
station["CHK_DATE"] = "2022-12-21"
print(f"After adding CHK_DATE: {station}")
print(f"Number of items: {len(station)}")

# Assigning to an existing key overwrites its value, no new item is added
station["QUAL"] = "Very Good"
print(f"After updating QUAL: {station}")
print(f"Number of items: {len(station)}")
```

You can also start with an empty dictionary and add items one at a time.

```python
# Start with an empty dictionary
new_station = {}
print(f"new_station: {new_station}")

# Add items one at a time
new_station["NAME"] = "Elk 2"
new_station["QUAL"] = "Excellent"
new_station["CHK_BY"] = "B Stewart"
print(f"After adding three items: {new_station}")
```

### 3. Checking for keys with `in` and `.get()`

Accessing a key that does not exist raises an error called a `KeyError`. Keys are case-sensitive, so `"qual"` and `"QUAL"` are two different keys.

```python
# Accessing a key that does not exist raises an error
# print(station["ELEV"])
# KeyError: 'ELEV'

# Keys are case-sensitive, so this raises an error too
# print(station["qual"])
# KeyError: 'qual'
```

The `in` operator checks whether a key exists in a dictionary and returns a boolean, `True` or `False`, the same way it checks whether an item is in a list. With a dictionary, `in` checks the keys only, not the values.

```python
# We will use variable station from Section 1
print(f"station: {station}")

# in checks whether a key exists in the dictionary
has_qual = "QUAL" in station
print(f"Is QUAL a key in station? {has_qual}")                # True

has_elev = "ELEV" in station
print(f"Is ELEV a key in station? {has_elev}")                # False

# Keys are case-sensitive
has_lower_qual = "qual" in station
print(f"Is qual a key in station? {has_lower_qual}")          # False

# in checks keys only, not values
has_name_value = "Bear Butte 1" in station
print(f"Is Bear Butte 1 a key in station? {has_name_value}")  # False
```

Checking with `in` before accessing a key avoids a `KeyError`.

```python
# Variable stores a field name to look up
# Change this value and rerun the cell to test both branches
field = "ELEV"
print(f"field: {field}")

if field in station:
    print(f"\t{field}: {station[field]}")
else:
    print(f"\t{field} is not a key in station")
```

The `.get()` method returns the value for a key without raising an error when the key is missing. It returns `None`, Python's value for "nothing here", unless you give it a second value to return instead, called a **default**.

```python
# .get() returns the value when the key exists
qual = station.get("QUAL")
print(f"QUAL: {qual}")

# .get() returns None when the key does not exist, instead of an error
elev = station.get("ELEV")
print(f"ELEV: {elev}")

# A second value sets the default to return when the key does not exist
elev_or_default = station.get("ELEV", "No elevation recorded")
print(f"ELEV: {elev_or_default}")
```

### 4. Removing items

`del` removes the item for a given key. `.pop()` removes the item for a given key and returns its value, so you can store it in a variable, the same way `.pop()` works with lists. Both raise a `KeyError` if the key does not exist.

```python
# We will use variable station from Section 1
print(f"station: {station}")

# del removes the item for a given key
del station["CHK_DATE"]
print(f"After del: {station}")

# .pop() removes the item for a given key and returns its value
removed_checker = station.pop("CHK_BY")
print(f"Removed: {removed_checker}")
print(f"After pop: {station}")
```

### 5. Looping over dictionaries

The `.keys()`, `.values()`, and `.items()` methods return all the keys, all the values, or all the key-value pairs in a dictionary. Each item from `.items()` is a tuple of `(key, value)`. The results look like lists wrapped in a label such as `dict_keys(...)`, and a `for` loop can iterate over them the same way it iterates over a list.

```python
# We will use variable station from Section 1
print(f"Keys: {station.keys()}")
print(f"Values: {station.values()}")
print(f"Items: {station.items()}")
```

Looping over `.keys()` gives you one key per pass through the loop. Use the key inside square brackets to get its value.

```python
for key in station.keys():
    print(f"\t{key}: {station[key]}")
```

Looping over a dictionary directly also loops over its keys, so this produces the same output as the loop above. This is the shorter form you will see most often in existing code and AI-generated code.

```python
for key in station:
    print(f"\t{key}: {station[key]}")
```

Looping over `.items()` gives you one `(key, value)` tuple per pass. Writing two variables after `for` separates the tuple into the key and the value, the same way `enumerate()` gives you an index and an item.

```python
for key, value in station.items():
    print(f"\t{key}: {value}")
```

A loop with an `if` inside can find and update specific values. Updating the value of an existing key inside a loop is fine. Adding or removing keys while looping over a dictionary raises a `RuntimeError`, so only update values inside the loop.

```python
# Dictionary stores quality ratings for each station
# Two ratings have inconsistent capitalization
station_quals = {
    "Elk 1": "Fair",
    "Bear Butte 1": "excellent",
    "Bear Butte 2": "Fair",
    "Whitewood 1": "very good",
}
print(f"station_quals: {station_quals}")

for name in station_quals.keys():
    qual = station_quals[name]
    # Check whether the rating is not already in title case
    if qual != qual.title():
        station_quals[name] = qual.title()  # Update the value for this key
        print(f"\tFixed {name}: {qual} -> {station_quals[name]}")

print(f"After fixing: {station_quals}")
```
