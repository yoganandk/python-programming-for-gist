# Loops

A GIS workflow often needs to repeat the same action many times, checking every station in a monitoring network, buffering every feature in a shapefile, running a tool across a list of files. Loops are how Python repeats a block of code without writing it out by hand for every item.

🎯 What You'll Learn

* Repeat code while a condition stays `True` using a `while` loop
* Iterate directly over a list using a `for` loop
* Exit a loop early with `break`
* Skip an iteration with `continue`
* Generate a sequence of numbers with `range()`
* Track iterations inside a loop
* See how a loop can be nested inside another loop

### 1. The `while` Loop

A `while` loop repeats a block of code for as long as a **condition** stays `True`. This condition is similar to ones used in conditional statements, but it is checked again after every pass through the loop's code block.  
A colon (`:`) follows the condition, and every line that belongs to the loop's code block is indented underneath it.

```python
# Example 1
# Variable stores the starting buffer radius in meters
buffer_radius_m = 5000
print(f"Before Loop: buffer_radius_m: {buffer_radius_m}")

# Keep buffering while the radius is at or below the maximum of 20000
while buffer_radius_m <= 20000:
    print(f"\tBuffer radius: {buffer_radius_m} meters")

    # Increment the radius by 5000 meters for the next buffer
    buffer_radius_m += 5000

# The buffer_radius_m variable still holds its last value after the loop ends
print(f"After Loop: buffer_radius_m: {buffer_radius_m}")
```

If the condition is `False` before the loop is ever reached, the loop's code block does not run at all, not even once.

```python
# Variable stores the starting buffer radius in meters
buffer_radius_m = 25000
print(f"Before Loop: buffer_radius_m: {buffer_radius_m}")

# Keep buffering while the radius is at or below the maximum of 20000
while buffer_radius_m <= 20000:
    print(f"\tBuffer radius: {buffer_radius_m} meters")

    # Increment the radius by 5000 meters for the next buffer
    buffer_radius_m += 5000

print(f"After Loop: buffer_radius_m: {buffer_radius_m}")
```

```python
# Example 2
# List stores buffer radii still waiting to be processed, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

# Reverse the list so .pop() removes items in the original order
buffer_radii_m.reverse()

# Keep looping while the list still has items in it 
# (a non-empty list is truthy, an empty list is falsy)
while buffer_radii_m:
    # Removes and returns the last item in the list
    radius = buffer_radii_m.pop()

    print(f"\tProcessing buffer radius: {radius} meters")

# The list is now empty, so the loop's condition is False
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")
```

### 2. The `for` Loop

A `for` loop **iterates** directly over each item in a collection like a list. Python advances through the list automatically, without you needing to manage a counter or a condition yourself.  
Each pass through the loop stores the current item in a **loop variable**.  
A colon (`:`) follows the loop's setup line, and its code block is indented underneath it.

```python
# List stores buffer radii to process this week, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

print("Begin buffering")
# for loop iterates directly over each item in the list
# radius is called the loop variable
for radius in buffer_radii_m:
    # The loop variable radius stores the current item
    # for this pass of the loop
    print(f"\tBuffer radius: {radius} meters")

# The loop variable radius still holds its last value after the loop ends
print(f"After Loop: Radius stored in loop variable: {radius}m")

# The original list remains intact
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")
```

### 3. The `break` Statement

A `while` or `for` loop normally keeps running until its condition becomes `False` or its list is exhausted. `break` stops the loop immediately, skipping any remaining passes.

```python
# Example 1 - while loop
# Variable stores the starting buffer radius in meters
buffer_radius_m = 5000
print(f"Before Loop: buffer_radius_m: {buffer_radius_m}")

# Keep buffering while the radius is at or below the maximum of 20000
while buffer_radius_m <= 20000:
    print(f"\tBuffer radius: {buffer_radius_m} meters")

    # Check if the radius is 15000
    if buffer_radius_m == 15000:
        # Condition is true, stop the loop
        print(f"\t\tStopping buffering at radius {buffer_radius_m}m.")

        break  # break out of the while loop

    # Increment the radius by 5000 meters for the next buffer
    buffer_radius_m += 5000

# The buffer_radius_m variable still holds its last value after the loop ends
print(f"After Loop: buffer_radius_m: {buffer_radius_m}")
```

```python
# Example 2 - for loop
# List stores buffer radii to process this week, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

print("Begin buffering")
# for loop iterates directly over each item in the list
# radius is called the loop variable
for radius in buffer_radii_m:
    # The loop variable radius stores the current item
    # for this pass of the loop
    print(f"\tBuffer radius: {radius} meters")

    # Check if the radius is 15000
    if radius == 15000:
        # Condition is true, stop the loop
        print(f"\t\tStopping buffering at radius {radius}m.")

        break  # break out of the for loop

# The loop variable radius still holds its last value after the loop ends
print(f"After Loop: Radius stored in loop variable: {radius}m")

# The original list remains intact
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")
```

### 4. The `continue` Statement

`continue` skips the rest of the current pass through the loop and moves on to the next one, without stopping the loop entirely the way `break` does.

```python
# Example 1 - while loop
# Variable stores the starting buffer radius in meters
buffer_radius_m = 5000
print(f"Before Loop: buffer_radius_m: {buffer_radius_m}")

while buffer_radius_m <= 20000:
    # Check if the radius is 15000
    if buffer_radius_m == 15000:
        print(f"\t\tSkipping this radius: {buffer_radius_m} meters")

        # Must increment here too, continue skips the increment
        # below, so without this line the loop would run forever
        buffer_radius_m += 5000

        continue  # continue to the next iteration of the while loop

    print(f"\tBuffer radius: {buffer_radius_m} meters")

    # Increment the radius by 5000 meters for the next buffer
    buffer_radius_m += 5000

print(f"After Loop: buffer_radius_m: {buffer_radius_m}")
```

```python
# Example 2 - for loop
# List stores buffer radii to process this week, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

for radius in buffer_radii_m:
    # Check if the radius is 15000
    if radius == 15000:
        print(f"\t\tSkipping this radius: {radius} meters")

        continue  # continue to the next iteration of the for loop

    print(f"\tBuffer radius: {radius} meters")

# The loop variable radius still holds its last value after the loop ends
print(f"After Loop: Radius stored in loop variable: {radius}m")

# The original list remains intact
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")
```

### 5. The `range()` Function

Sometimes, while using a `for` loop, you may want to repeat something a set number of times without an existing list to loop over. The `range()` function generates a sequence of numbers you can loop over directly.

`range(stop)` generates numbers starting at 0, up to but not including `stop`.

```python
# range(6) generates numbers 0, 1, 2, 3, 4 and 5
for i in range(6):
    # Loop Variable i stores the current number for this pass
    print(f"Loop variable i: {i}")
```

`range(start, stop)` generates numbers starting at `start`, up to but not including `stop`.

```python
# range(100, 106) generates numbers 100, 101, 102, 103, 104 and 105
for i in range(100, 106):
    # Loop Variable i stores the current number for this pass
    print(f"Loop variable i: {i}")
```

`range(start, stop, step)` adds a third value that controls how much each number increases by. A positive `step` counts up.

```python
# range(5000, 20000, 5000) generates numbers 5000, 10000 and 15000
# increment is 5000
for i in range(5000, 20000, 5000):
    # Loop Variable i stores the current number for this pass
    print(f"Loop variable i: {i}")
```

A negative `step` counts down instead.

```python
# range(20000, 5000, -5000) generates numbers 20000, 15000 and 10000
# decrement is 5000
for i in range(20000, 5000, -5000):
    # Loop Variable i stores the current number for this pass
    print(f"Loop variable i: {i}")
```

### 6. Tracking Iterations

A loop's condition or its loop variable tells you what value you are on, but not how many times the loop has run, or which position in a collection you are at.  
One way to track that is a **counter** variable you manage yourself, in either a `while` or a `for` loop.  
For a `for` loop specifically, `enumerate()` offers a cleaner way to get the same information without managing the counter by hand.

```python
# Example 1 - while loop
# Variable stores the starting buffer radius in meters
buffer_radius_m = 5000
print(f"Before Loop: buffer_radius_m: {buffer_radius_m}")

# Variable tracks how many buffers have been processed
buffer_count = 0
print(f"Before Loop: buffer_count: {buffer_count}")

# Keep buffering while the radius is at or below the maximum of 20000
while buffer_radius_m <= 20000:
    # Increment the counter for this iteration
    buffer_count += 1
    print(f"\tCount: {buffer_count} | Buffer radius: {buffer_radius_m} meters")

    # Increment the radius by 5000 meters for the next buffer
    buffer_radius_m += 5000

# The buffer_radius_m variable still holds its last value after the loop ends
print(f"After Loop: buffer_radius_m: {buffer_radius_m}")

# The buffer_count variable still holds its last value after the loop ends
print(f"After Loop: Total Iterations or buffer_count: {buffer_count}")
```

```python
# Example 2 - for loop
# List stores buffer radii to process this week, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

# Variable tracks how many buffers have been processed
buffer_count = 0
print(f"Before Loop: buffer_count: {buffer_count}")

# for loop iterates directly over each item in the list
# radius is called the loop variable
for radius in buffer_radii_m:
    # Increment the counter for this iteration
    buffer_count += 1
    print(f"\tCount: {buffer_count} | Buffer radius: {radius} meters")

# The loop variable radius still holds its last value after the loop ends
print(f"After Loop: Radius stored in loop variable: {radius}m")

# The original list remains intact
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")

# The buffer_count variable still holds its last value after the loop ends
print(f"After Loop: Total Iterations or buffer_count: {buffer_count}")
```

When using `for` loops, the `enumerate()` function pairs each item in a list with a count, starting at 0 by default, which removes the need to manage a counter variable yourself.

```python
# List stores buffer radii to process this week, in meters
buffer_radii_m = [5000, 10000, 15000, 20000]
print(f"Before Loop: buffer_radii_m: {buffer_radii_m}")

# start=1 in enumerate() makes the count begin at 1 instead of the default 0
for count, radius in enumerate(buffer_radii_m, start=1):
    print(f"\tCount: {count} | Buffer radius: {radius} meters")

# The loop variable radius still holds its last value after the loop ends
print(f"After Loop: Radius stored in loop variable: {radius}m")

# The original list remains intact
print(f"After Loop: buffer_radii_m: {buffer_radii_m}")

# The count variable still holds its last value after the loop ends
print(f"After Loop: Total Iterations: {count}")
```

### 7. Nested Loops

A loop can contain another loop inside it, this is called a **nested** loop. The inner loop runs completely, start to finish, for every single pass of the outer loop.  
Nesting is not limited to two loops, a loop can be nested inside another loop that is itself nested inside a third, and so on, though it gets harder to follow with each additional level.

```python
# List stores station names to buffer
station_names = ["Elk 1", "Bear Butte 1", "Bear Butte 2"]

# List stores buffer radii to create around each station, in meters
buffer_radii_m = [5000, 10000, 15000]

for s_count, station in enumerate(station_names, start=1):
    # Outer loop: one pass per station
    print(f"Outer Loop: s_count: {s_count} | Station: {station}")
    for r_count, radius in enumerate(buffer_radii_m, start=1):
        # Inner loop: one pass per radius for each station
        print(f"\tInner Loop: Station: {station} | r_count: {r_count} "
              f"| Radius: {radius} meters")

# Note: This demonstrates a for loop nested inside a for loop. However,
# you can nest any combination of while and/or for loops.
```
